## 1. Arti Singkatan `pmbt` pada Catatan 05

### Pertanyaan
> *"apa yang pmbt?"*

### Jawaban & Penjelasan
Dalam catatan materi, **`pmbt`** adalah singkatan dari **pembayaran** (*payment* / arus kas).

Singkatan ini digunakan pada diagram garis waktu (*timeline diagram*) dan tabel perbandingan untuk menandai titik pembayaran:
- **`pmbt-1`** : Pembayaran pertama / pembayaran ke-1.
- **`pmbt-n`** : Pembayaran terakhir / pembayaran ke-$n$.

---

### Konteks Titik Evaluasi Nilai Waktu (*Valuation Timing*)

Posisi waktu pembayaran menentukan bagaimana nilai sekarang ($PV$) dan nilai akumulasi ($AV$) dievaluasi:

| Karakteristik | Anuitas Akhir (*Annuity-Immediate*) | Anuitas Awal (*Annuity-Due*) |
| :--- | :--- | :--- |
| **Waktu Pembayaran** | $t = 1, 2, \dots, n$ | $t = 0, 1, \dots, n-1$ |
| **Titik Evaluasi $PV$** | $t = 0$ (**1 periode sebelum pmbt-1**) | $t = 0$ (**pada saat pmbt-1**) |
| **Rumus Nilai Sekarang ($PV$)** | $a_{\overline{n}\vert i} = \dfrac{1 - v^n}{i}$ | $\ddot{a}_{\overline{n}\vert i} = \dfrac{1 - v^n}{d}$ |
| **Titik Evaluasi $AV$** | $t = n$ (**pada saat pmbt-$n$**) | $t = n$ (**1 periode setelah pmbt-$n$**) |
| **Rumus Terakumulasi ($AV$)** | $s_{\overline{n}\vert i} = \dfrac{(1+i)^n - 1}{i}$ | $\ddot{s}_{\overline{n}\vert i} = \dfrac{(1+i)^n - 1}{d}$ |

---

### Hubungan Fundamental Antar Anuitas

Karena pembayaran anuitas-due terjadi tepat 1 periode lebih awal dibandingkan anuitas-immediate, uang tersebut memiliki waktu tambahan 1 periode untuk memupuk bunga:

$$\ddot{a}_{\overline{n}\vert i} = (1+i) a_{\overline{n}\vert i} = 1 + a_{\overline{n-1}\vert i}$$

$$\ddot{s}_{\overline{n}\vert i} = (1+i) s_{\overline{n}\vert i} = s_{\overline{n+1}\vert i} - 1$$

dengan faktor diskon $v = (1+i)^{-1}$ dan tingkat diskon $d = \dfrac{i}{1+i} = 1 - v$.

## 2. Mengapa Nilai Akumulasi Sebesar $1 \cdot (1+i)^k$? Kenapa Dikali $(1+i)$?

### Pertanyaan
> *"kenapa nilai akumulasi adalah 1*(1+i)^k? Kenapa dikali (1+i)?"*

### Jawaban & Penjelasan

#### 1. Dari Mana Asal Faktor $(1+i)$? (Akumulasi 1 Periode)
Misalkan Anda menabung modal awal sebesar pokok $P = 1$ dengan suku bunga efektif $i$ per periode:
- Di akhir periode 1, Anda tetap memiliki **modal pokok** sebesar $1$.
- Ditambah **bunga** yang dihasilkan modal tersebut sebesar $1 \times i = i$.

Total saldo (nilai akumulasi) di akhir periode 1 adalah gabungan pokok dan bunga:
$$\text{Saldo Akhir} = \text{Pokok} + \text{Bunga} = 1 + i = 1 \cdot (1+i)$$

Faktor $(1+i)$ adalah hasil pemfaktoran aljabar:
- Angka $1$ mewakili pengembalian modal pokok $100\%$.
- Suku $+i$ mewakili tambahan keuntungan bunga.

---

#### 2. Mengapa Menjadi Pangkat $k$ (Bunga Berbunga / *Compounding*)?
Dalam sistem bunga majemuk, di periode berikutnya bunga dihitung dari **seluruh saldo terakhir** (pokok + bunga periode sebelumnya yang diinvestasikan kembali):

- **Akhir Periode 1**:
  $$\text{Saldo}_1 = 1 \cdot (1+i)$$
- **Akhir Periode 2** (Saldo awal periode 2 adalah $(1+i)$):
  $$\text{Saldo}_2 = \text{Saldo}_1 + \text{Saldo}_1 \cdot i = (1+i) \cdot (1+i) = (1+i)^2$$
- **Akhir Periode 3**:
  $$\text{Saldo}_3 = \text{Saldo}_2 \cdot (1+i) = (1+i)^2 \cdot (1+i) = (1+i)^3$$

Secara umum, setiap melewati **$1$ periode**, saldo dikalikan dengan faktor $(1+i)$.
Jika uang dibiarkan mengendap selama **$k$ periode**, maka kita mengalikan faktor akumulasi tersebut sebanyak $k$ kali:

$$AV = 1 \cdot \underbrace{(1+i) \times (1+i) \times \dots \times (1+i)}_{k \text{ kali}} = 1 \cdot (1+i)^k$$

---

#### 3. Penerapan pada Anuitas (Catatan 05)
Jika pembayaran sebesar $1$ dilakukan pada waktu $t$, lalu kita ingin mengetahui nilainya di masa depan pada titik waktu evaluasi $T$ (di mana $T \ge t$):
- Jarak waktu uang tersebut mengendap dan memupuk bunga adalah $k = T - t$ periode.
- Maka nilai terakumulasi dari pembayaran tersebut pada waktu $T$ adalah:
  $$AV = 1 \cdot (1+i)^{T - t}$$

**Contoh pada Anuitas Akhir ($s_{\overline{n}\vert i}$):** Evaluasi dilakukan pada $t = n$:
- Pembayaran ke-1 (pada $t=1$) mengendap selama $n - 1$ periode $\rightarrow 1 \cdot (1+i)^{n-1}$
- Pembayaran ke-2 (pada $t=2$) mengendap selama $n - 2$ periode $\rightarrow 1 \cdot (1+i)^{n-2}$
- $\dots$
- Pembayaran ke-$n$ (pada $t=n$) mengendap selama $n - n = 0$ periode $\rightarrow 1 \cdot (1+i)^0 = 1$

Sehingga total nilai akumulasi seluruh arus kas adalah:
$$s_{\overline{n}\vert i} = 1 + (1+i) + (1+i)^2 + \dots + (1+i)^{n-1} = \dfrac{(1+i)^n - 1}{i}$$

## 3. Mengapa Faktor Diskonto ($v$) adalah Resiprokal dari $(1+i)$?

### Pertanyaan
> *"kenapa faktor diskonto adalah resiprokal dari 1+i?"*

### Jawaban & Penjelasan

#### 1. Konsep Pembalik Waktu (*Time Inversion*)
Faktor diskonto ($v$) pada hakikatnya adalah **operasi kebalikan (invers)** dari faktor akumulasi $(1+i)$:

- **Akumulasi (Maju ke Masa Depan)** menjawab pertanyaan:
  > *"Jika saya menginvestasikan $\$1$ sekarang ($t=0$), berapa nilainya $1$ periode ke depan ($t=1$)?"*
  $$\text{Nilai di } t=1 = 1 \times (1+i)$$

- **Diskonto (Mundur ke Masa Sekarang)** menjawab pertanyaan kebalikannya:
  > *"Berapa modal awal $P$ yang harus saya investasikan sekarang ($t=0$) agar tepat bernilai $\$1$ pada $1$ periode ke depan ($t=1$)?"*

---

#### 2. Penurunan Aljabar
Dari hubungan akumulasi, nilai modal $P$ setelah $1$ periode harus sama dengan $1$:
$$P \cdot (1+i) = 1$$

Untuk mencari nilai $P$, bagi kedua ruas dengan $(1+i)$:
$$P = \frac{1}{1+i} = (1+i)^{-1}$$

Besaran modal $P$ inilah yang dinamakan **faktor diskonto ($v$)**:
$$v = \frac{1}{1+i}$$

---

#### 3. Simetri Nilai Waktu dari Uang (*Time Value of Money*)

| Gerakan Garis Waktu | Operasi Matematika | Pengali |
| :--- | :--- | :--- |
| **Maju 1 periode ($t \to t+1$)** | Mengalikan dengan faktor akumulasi | $\times (1+i)$ |
| **Mundur 1 periode ($t+1 \to t$)** | Membagi dengan $(1+i)$ (mengalikan inversnya) | $\times \dfrac{1}{1+i} = \times v$ |
| **Maju $k$ periode** | Dikalikan berulang $k$ kali | $\times (1+i)^k$ |
| **Mundur $k$ periode** | Dibagi berulang $k$ kali | $\times v^k = \times (1+i)^{-k}$ |

---

#### 4. Penerapan pada Nilai Sekarang Anuitas ($a_{\overline{n}\vert i}$)
Setiap pembayaran $\$1$ pada waktu $t$ bernilai $1 \cdot v^t$ pada saat $t=0$.
Menjumlahkan seluruh $n$ pembayaran menghasilkan nilai sekarang:
$$a_{\overline{n}\vert i} = v^1 + v^2 + v^3 + \dots + v^n = \sum_{t=1}^n \frac{1}{(1+i)^t} = \frac{1 - v^n}{i}$$

## 4. Latihan Soal 05 (Soal 2): Mengapa Rumus Akumulasi Memakai Tingkat Diskon ($d$) Bukan Suku Bunga ($i$)?

### Pertanyaan
> *"Pada latihan soal '05 Annuitas dan Perpetuitas 1\tugas-dan-latihan\latihan_soal_05.pdf' soal nomor 2 kenapa memakai di diskon bukan suku bunga?"*

### Jawaban & Penjelasan

#### 1. Identifikasi Jenis Transaksi: Anuitas Awal (*Annuity-Due*)
Perhatikan bunyi dan garis waktu pada **Soal 2**:
- Setoran sebesar $R$ dilakukan pada **setiap awal tahun** sebanyak 18 kali setoran, yaitu pada $t = 0, 1, 2, \dots, 17$.
- Target akumulasi dana sebesar $\$80{,}000$ dievaluasi tepat pada saat anak berusia 18 tahun ($t = 18$).

**Poin Kunci Garis Waktu:**
- Setoran terakhir terjadi di $t = 17$.
- Nilai target dihitung pada $t = 18$ (**tepat 1 tahun / 1 periode setelah setoran terakhir**).
- Karena titik evaluasi akumulasi berada 1 periode setelah setoran terakhir, transaksi ini merupakan **Anuitas Awal (*Annuity-Due*)**, sehingga simbol aktuarial yang digunakan adalah $\mathbf{\ddot{s}_{\overline{n}\vert i}}$ (*s double dot*), bukan $s_{\overline{n}\vert i}$.

---

#### 2. Penurunan Aljabar: Mengapa Muncul $d$ di Penyebut?
Mari kita akumulasikan seluruh 18 setoran ke titik evaluasi $t = 18$:
- Setoran ke-1 ($t=0$) berbunga selama 18 tahun $\longrightarrow R(1+i)^{18}$
- Setoran ke-2 ($t=1$) berbunga selama 17 tahun $\longrightarrow R(1+i)^{17}$
- $\dots$
- Setoran ke-18 ($t=17$) berbunga selama 1 tahun $\longrightarrow R(1+i)^1$

Total akumulasi nilai seluruh setoran:
$$\text{AV} = R \cdot \underbrace{\left[ (1+i)^1 + (1+i)^2 + \dots + (1+i)^{18} \right]}_{\ddot{s}_{\overline{18}\vert i}}$$

Faktorkan keluar $(1+i)$:
$$\ddot{s}_{\overline{18}\vert i} = (1+i) \cdot \left[ 1 + (1+i) + (1+i)^2 + \dots + (1+i)^{17} \right]$$

Deret di dalam kurung siku adalah deret geometri anuitas akhir biasa ($s_{\overline{18}\vert i}$):
$$\ddot{s}_{\overline{18}\vert i} = (1+i) \cdot s_{\overline{18}\vert i} = (1+i) \cdot \left[ \frac{(1+i)^{18} - 1}{i} \right]$$

Pindahkan pengali $(1+i)$ ke bagian penyebut pecahan:
$$\ddot{s}_{\overline{18}\vert i} = \frac{(1+i)^{18} - 1}{\dfrac{i}{1+i}}$$

Ingat hubungan fundamental antara tingkat bunga efektif $i$ dan tingkat diskon efektif $d$:
$$d = \frac{i}{1+i} = 1 - v \quad \Longleftrightarrow \quad \frac{1}{d} = \frac{1+i}{i}$$

Sehingga penyebut $\dfrac{i}{1+i}$ disederhanakan menjadi **$d$**:
$$\mathbf{\ddot{s}_{\overline{18}\vert i} = \frac{(1+i)^{18} - 1}{d}}$$

---

#### 3. Pola Pembagi pada Notasi Standar Aktuaria

| Jenis Anuitas | Nilai Sekarang ($PV$) | Nilai Terakumulasi ($AV$) | Pembagi Standar |
| :--- | :---: | :---: | :---: |
| **Anuitas Akhir (*Immediate*)** | $a_{\overline{n}\vert i} = \dfrac{1 - v^n}{i}$ | $s_{\overline{n}\vert i} = \dfrac{(1+i)^n - 1}{i}$ | **$i$** (Suku Bunga) |
| **Anuitas Awal (*Due*)** | $\ddot{a}_{\overline{n}\vert i} = \dfrac{1 - v^n}{d}$ | $\ddot{s}_{\overline{n}\vert i} = \dfrac{(1+i)^n - 1}{d}$ | **$d$** (Tingkat Diskon) |

> **Catatan Penting:** Soal ini tetap menggunakan suku bunga $i = 5.5\%$. Munculnya $d$ di penyebut hanyalah penyederhanaan aljabar dari $(1+i)/i$.

---

#### 4. Tiga Cara Ekivalen Menghitung $\ddot{s}_{\overline{18}\vert 0.055}$
Ketiga formula berikut menghasilkan nilai yang **100% identik**:
1. **Bentuk Baku Diskon ($d$):**
   $$d = \frac{0.055}{1.055} \approx 0.0521327 \implies \ddot{s}_{\overline{18}\vert 0.055} = \frac{(1.055)^{18} - 1}{0.0521327} \approx 31.1030$$
2. **Bentuk Pengali Faktor Akumulasi $(1+i)$:**
   $$\ddot{s}_{\overline{18}\vert 0.055} = (1.055) \cdot s_{\overline{18}\vert 0.055} = (1.055) \times \frac{(1.055)^{18} - 1}{0.055} \approx 31.1030$$
3. **Bentuk Pergeseran Deret $(s_{\overline{n+1}\vert} - 1)$:**
   $$\ddot{s}_{\overline{18}\vert 0.055} = s_{\overline{19}\vert 0.055} - 1 = \frac{(1.055)^{19} - 1}{0.055} - 1 \approx 31.1030$$

Maka besarnya setoran tahunan adalah:
$$R = \frac{80{,}000}{\ddot{s}_{\overline{18}\vert 0.055}} = \frac{80{,}000}{31.1030} = \mathbf{\$2{,}572.09 \text{ per tahun}}$$

## 6. Mengapa Rumus Nilai Sekarang adalah $R \cdot a$ dan Nilai Akumulasi adalah $R \cdot s$?

### Pertanyaan
> *"kenapa rumus present value adalah R*a dan future value R*s"*

### Jawaban & Penjelasan

Alasan mendasar mengapa rumus *Present Value* adalah $R \cdot a$ dan *Future Value* adalah $R \cdot s$ terletak pada **definisi simbol baku aktuaria** dan **sifat linearitas nilai waktu dari uang (*Time Value of Money*)**.

---

#### 1. Definisi Simbol Baku Aktuaria ($a$ dan $s$) Adalah Nilai Per Unit ($1$ Satuan Uang)
Dalam matematika aktuaria, simbol $a_{\overline{n}\vert i}$ dan $s_{\overline{n}\vert i}$ didefinisikan khusus untuk rangkaian pembayaran sebesar **$1$ unit** (misalnya $\$1$ atau Rp1) per periode:
- **$a_{\overline{n}\vert i}$** = *Present Value* (Nilai Sekarang) dari serangkaian pembayaran **$\$1$** sebanyak $n$ kali.
- **$s_{\overline{n}\vert i}$** = *Future / Accumulated Value* (Nilai Terakumulasi) dari serangkaian pembayaran **$\$1$** sebanyak $n$ kali.

*(Asal kata notasi: **$a$** dari bahasa Latin *annuitas* / nilai sekarang, sedangkan **$s$** dari kata *summe / sum* yaitu akumulasi jumlah akhir).*

---

#### 2. Penurunan Aljabar & Sifat Perkalian Skalar ($R$)

Jika setiap periode besar pembayaran bukan $\$1$, melainkan sejumlah konstan **$R$** (misalnya $R = \$2{,}572.09$):

##### A. Penurunan Nilai Sekarang ($PV = R \cdot a_{\overline{n}\vert i}$)
Setiap pembayaran $R$ ditarik mundur (didiskontokan) ke titik evaluasi sekarang ($t=0$):
$$\text{PV} = R \cdot v^1 + R \cdot v^2 + R \cdot v^3 + \dots + R \cdot v^n$$

Faktorkan keluar konstanta $R$:
$$\text{PV} = R \cdot \underbrace{\left( v^1 + v^2 + v^3 + \dots + v^n \right)}_{a_{\overline{n}\vert i}} = \mathbf{R \cdot a_{\overline{n}\vert i}}$$

##### B. Penurunan Nilai Akumulasi / Masa Depan ($FV = R \cdot s_{\overline{n}\vert i}$)
Setiap pembayaran $R$ diakumulasikan maju dengan bunga berbunga ke titik akhir ($t=n$):
$$\text{FV} = R \cdot (1+i)^{n-1} + R \cdot (1+i)^{n-2} + \dots + R \cdot (1+i)^0$$

Faktorkan keluar konstanta $R$:
$$\text{FV} = R \cdot \underbrace{\left[ 1 + (1+i)^1 + (1+i)^2 + \dots + (1+i)^{n-1} \right]}_{s_{\overline{n}\vert i}} = \mathbf{R \cdot s_{\overline{n}\vert i}}$$

---

#### 3. Hubungan Ekuivalensi Fundamental antara $PV$ dan $FV$

Karena nilai masa depan ($FV$) adalah nilai sekarang ($PV$) yang diakumulasikan penuh selama $n$ periode:
$$\text{FV} = \text{PV} \cdot (1+i)^n$$

Substitusikan $\text{PV} = R \cdot a_{\overline{n}\vert i}$ dan $\text{FV} = R \cdot s_{\overline{n}\vert i}$:
$$R \cdot s_{\overline{n}\vert i} = (R \cdot a_{\overline{n}\vert i}) \cdot (1+i)^n \quad \Longleftrightarrow \quad \mathbf{s_{\overline{n}\vert i} = a_{\overline{n}\vert i} \cdot (1+i)^n}$$

*(Prinsip perkalian dengan $R$ ini berlaku identik untuk Anuitas Awal: $\text{PV} = R \cdot \ddot{a}_{\overline{n}\vert i}$ dan $\text{AV} = R \cdot \ddot{s}_{\overline{n}\vert i}$).*

## 7. Latihan Soal 05 (Soal 4): Maksud dari Notasi Anuitas Ditunda ${}_{14}| a_{\overline{20}\vert}$

### Pertanyaan
> *"Apa maksud dari notasi 14|a_20 pada soal 4?"*

### Jawaban & Penjelasan

Notasi **${}_{14}| a_{\overline{20}\vert i}$** (sering ditulis secara ringkas sebagai `${}_{14}|a_{20}`) adalah **Notasi Standar Aktuaria Internasional** untuk **Anuitas Ditunda (*Deferred Annuity*)**.

---

#### 1. Pembongkaran Arti Simbol: ${}_{m}| a_{\overline{n}\vert i}$

Notasi ini dibaca sebagai gabungan dua komponen:
1. **Prefiks ${}_{14}|$ ($m = 14$ periode penundaan):**
   - Menunjukkan **masa penundaan (*deferral period*)** selama **14 tahun**.
   - Selama $t = 1, 2, \dots, 14$, **tidak ada pembayaran sama sekali** ($0$ arus kas).
2. **Sufiks $a_{\overline{20}\vert}$ ($n = 20$ periode pembayaran):**
   - Menunjukkan suatu **anuitas akhir biasa (*annuity-immediate*)** dengan durasi **20 kali pembayaran**.

> **Arti Lengkap:**
> Nilai sekarang pada saat ini ($t = 0$) dari serangkaian pembayaran tahunan sebesar $\$1$ sebanyak $20$ kali, yang pembayarannya **baru mulai aktif setelah ditunda selama 14 tahun**.

---

#### 2. Mengapa Penundaannya $14$ Tahun, Padahal Pembayaran Mulai di Usia 65 ($t = 15$)?

Perhatikan garis waktu (*timeline*) pada **Soal 4**:
- Tertanggung saat ini berusia 50 tahun ($t = 0$, titik evaluasi $PV_0$).
- Manfaat pensiun pertama cair tepat pada usia 65 tahun ($t = 15$).
- Total 20 pembayaran tahunan berlangsung dari $t = 15$ sampai $t = 34$ (usia 84 tahun).

$$\begin{array}{cccccc}
t = 0 & \dots & t = 14 & t = 15 & t = 16 & \dots & t = 34 \\
\text{(Usia 50)} & & \text{(Usia 64)} & \text{(Usia 65)} & \text{(Usia 66)} & & \text{(Usia 84)} \\
\hline
\text{Evaluasi } PV_0 & \longleftarrow & PV_{14} & \text{pmbt-1} & \text{pmbt-2} & \dots & \text{pmbt-20} \\
& \underbrace{\quad\quad\quad\quad}_{\text{Masa Tunda } (m=14)} & & (\$12{,}000) & (\$12{,}000) & \dots & (\$12{,}000)
\end{array}$$

**Alasan Kunci:**
- Pada rumus anuitas akhir $a_{\overline{n}\vert}$, titik evaluasi nilai sekarang selalu berada **1 periode sebelum pembayaran pertama**.
- Karena pembayaran pertama terjadi di $t = 15$, maka titik awal nilai sekarang 20 pembayaran pensiun tersebut berada di **$t = 14$** ($PV_{14} = R \cdot a_{\overline{20}\vert}$).
- Untuk membawa nilai $PV_{14}$ kembali ke hari ini ($t = 0$), kita mendiskontokannya sejauh **14 tahun**, sehingga notasinya adalah **${}_{14}| a_{\overline{20}\vert}$**.

---

#### 3. Dua Metode Matematis Menghitung ${}_{14}| a_{\overline{20}\vert}$

Kedua cara berikut menghasilkan nilai yang **100% identik**:

##### Metode 1: Mengalikan dengan Faktor Diskon $v^{14}$ (Metode Paling Praktis)
Hitung nilai tunai di $t = 14$, lalu diskontokan 14 periode ke $t = 0$:
$${}_{14}| a_{\overline{20}\vert i} = v^{14} \cdot a_{\overline{20}\vert i} = (1+i)^{-14} \cdot \left[ \frac{1 - (1+i)^{-20}}{i} \right]$$

Pada Soal 4 ($R = \$12{,}000$, $i = 6.5\%$):
$$PV_{14} = 12{,}000 \cdot a_{\overline{20}\vert 0.065} = 12{,}000 \times 11.018507 = \$132{,}222.08$$
$$PV_0 = \$132{,}222.08 \times (1.065)^{-14} = \mathbf{\$54{,}824.97}$$

##### Metode 2: Selisih Dua Anuitas (*Difference of Two Annuities*)
Bayangkan anuitas berjalan penuh selama 34 tahun ($t = 1$ s.d. $34$), lalu kurangkan dengan anuitas 14 tahun pertama ($t = 1$ s.d. $14$) yang tidak pernah dibayarkan:

$$\implies PV_0 = 12{,}000 \times (13.585573 - 9.017551) = 12{,}000 \times 4.568722 = \mathbf{\$54{,}824.97}$$

---

#### 4. Hubungan Ekivalen dengan Notasi Anuitas Awal ($\ddot{a}$)
Jika dipandang sebagai anuitas awal (di mana titik evaluasi tepat di saat setoran pertama $t = 15$), maka masa tundanya adalah 15 tahun:

## 9. Latihan Soal 05 (Soal 6): Perbedaan Konsep *Balloon Payment* vs *Drop Payment*

### Pertanyaan
> *"Saya sedang mengerjakan soal 6, apa itu balloon payment dan drop payment?"*

### Jawaban & Penjelasan

Ketika suatu pinjaman ($L$) dicicil dengan pembayaran tetap sebesar $R$ per periode, waktu pelunasan ($n$) sering kali menghasilkan angka pecahan / desimal (pada **Soal 6**: $n \approx 11.0028$ tahun).

Artinya, setelah **11 kali cicilan reguler penuh sebesar $\$4{,}000$**, masih terdapat sedikit sisa saldo utang (*outstanding balance*) $B_{11} > 0$ yang belum lunas (pada Soal 6: $B_{11} = \$11.16$).

Untuk melunasi sisa pecahan saldo tersebut, terdapat **2 opsi standar aktuarial & perbankan**:

---

#### 1. *Balloon Payment* (Pembayaran Menggelembung di Periode Terakhir)

- **Waktu Pembayaran:** Dibayarkan **bersamaan dengan cicilan reguler terakhir** (pada $t = 11$).
- **Mekanisme:** Cicilan reguler ke-11 digabungkan dengan seluruh sisa saldo utang yang tersisa.
- **Ciri Khas Nominal:** Nilai pembayaran menjadi **lebih besar dari cicilan biasa** ($> R$), seolah-olah pembayarannya "menggelembung" (*balloon*).
- **Formulasi & Perhitungan (Soal 6):**
  $$\text{Balloon Payment} = R + B_{11} = \$4{,}000 + \$11.16 = \mathbf{\$4{,}011.16 \quad (\text{di } t = 11)}$$
- **Total Transaksi:** Tetap berlangsung selama **11 kali pembayaran**.

---

#### 2. *Drop Payment* (Pembayaran Sisa Menetes di Periode Tambahan)

- **Waktu Pembayaran:** Dibayarkan **1 periode setelah cicilan reguler terakhir** (pada $t = 12$).
- **Mekanisme:** Pada $t = 11$, peminjam HANYA membayar cicilan normal sebesar $\$4{,}000$. Sisa saldo $B_{11}$ dibiarkan berjalan 1 tahun lagi (terkena bunga selama 1 tahun), lalu dilunasi penuh pada $t = 12$.
- **Ciri Khas Nominal:** Nilai pembayaran pada $t = 12$ menjadi **jauh lebih kecil / menurun drastis** ($< R$), seperti "tetesan sisa" (*drop*).
- **Formulasi & Perhitungan (Soal 6):**
  $$\text{Drop Payment } (X) = B_{11} \times (1 + i) = \$11.16 \times 1.07 = \mathbf{\$11.94 \quad (\text{di } t = 12)}$$
- **Total Transaksi:** Menjadi **12 kali pembayaran** (11 kali $\$4{,}000$ ditambah 1 kali $\$11.94$).

---

#### 3. Tabel Perbandingan Cepat

| Karakteristik | *Balloon Payment* | *Drop Payment* |
| :--- | :--- | :--- |
| **Waktu Pelunasan** | Pada periode ke-$k$ ($t = 11$) | Pada periode ke-$(k+1)$ ($t = 12$) |
| **Besar Pembayaran Terakhir** | $\mathbf{> R}$ (Lebih besar: $\$4{,}011.16$) | $\mathbf{< R}$ (Jauh lebih kecil: $\$11.94$) |
| **Jumlah Kali Bayar** | $k$ kali (11 kali) | $k + 1$ kali (12 kali) |
| **Rumus Pelunasan** | $P_{\text{balloon}} = R + B_k$ | $X_{\text{drop}} = B_k \cdot (1+i)$ |

## 10. Konsep Fundamental: Arti dan Fungsi Rumus $PV_0 \cdot (1+i)^n$

### Pertanyaan
> *"PV_0 * (1+i^n) itu rumus apa?"*

### Jawaban & Penjelasan

Rumus **$PV_0 \cdot (1+i)^n$** (maksudnya $PV_0 \times (1+i)^n$) adalah **Rumus Akumulasi Nilai Waktu dari Uang (*Compound Interest Accumulation Formula*)** atau rumus untuk **mengubah Nilai Sekarang (*Present Value*) menjadi Nilai Terakumulasi di Masa Depan (*Future / Accumulated Value*)**.

---

#### 1. Makna Matematis & Fisis

Faktor **$(1+i)^n$** disebut sebagai **faktor akumulasi (*accumulation factor*)** selama $n$ periode.

- Jika Anda memiliki sejumlah nilai ekuivalen pada saat ini ($t = 0$), yaitu **$PV_0$**.
- Lalu Anda membiarkan nilai tersebut mengendap dan memupuk bunga majemuk sebesar $i$ per periode selama **$n$ periode** ke depan.
- Maka nilai uang tersebut saat tiba di titik waktu masa depan ($t = n$) adalah:
  $$\mathbf{AV_n = FV_n = PV_0 \cdot (1+i)^n}$$

---

#### 2. Penerapan Rumus Ini pada Berbagai Konteks Materi

##### A. Kas Tunggal (*Lump Sum / Single Deposit*)
Jika Anda menabung modal awal sebesar $P$ di bank pada $t=0$, maka total saldo tabungan di akhir tahun ke-$n$ adalah:
$$S = P \cdot (1+i)^n$$

##### B. Hubungan antara Present Value Anuitas ($a$) dan Future Value Anuitas ($s$)
Jika seluruh $n$ pembayaran anuitas telah didiskontokan ke titik awal ($PV_0 = R \cdot a_{\overline{n}\vert i}$), kita dapat langsung mencari nilai akumulasinya di titik akhir ($t = n$) tanpa menghitung deret dari awal lagi:
$$AV_n = PV_0 \cdot (1+i)^n = \left( R \cdot a_{\overline{n}\vert i} \right) \cdot (1+i)^n = R \cdot s_{\overline{n}\vert i}$$

##### C. Anuitas Ditunda (*Deferred Annuity* - Contoh Soal 4)
Pada Soal 4, jika diketahui nilai sekarang manfaat pensiun hari ini adalah $PV_0 = \$54{,}753.20$ (di usia 50 tahun / $t=0$):
- Nilai saldo saat memasuki usia 64 ($t=14$): $PV_{14} = PV_0 \cdot (1+i)^{14} = \$132{,}222.08$
- Nilai saldo jika diakumulasi sampai usia 84 ($t=34$): $AV_{34} = PV_0 \cdot (1+i)^{34} = \$465{,}903.72$

##### D. Sisa Saldo Utang (*Retrospective Balance* - Contoh Soal 6)
Nilai pinjaman awal $L$ yang bertumbuh dengan bunga setelah $k$ periode dihitung dengan:
$$\text{Nilai Pinjaman Terakumulasi} = L \cdot (1+i)^k$$

---

#### 3. Ringkasan Arah Gerak Waktu (*Time Shift*)

| Arah Perpindahan Waktu | Operasi Matematika | Rumus |
| :--- | :--- | :---: |
| **Maju ke Masa Depan ($t=0 \longrightarrow t=n$)** | **Mengalikan** dengan faktor akumulasi | $\mathbf{FV = PV \cdot (1+i)^n}$ |
| **Mundur ke Masa Sekarang ($t=n \longrightarrow t=0$)** | **Membagi** dengan $(1+i)^n$ (mengalikan $v^n$) | $\mathbf{PV = FV \cdot (1+i)^{-n} = FV \cdot v^n}$ |

## 11. Latihan Soal 05 (Soal 8): Mengapa Rumus Verifikasi Berjalan, dan Apakah $PV$ Sama dengan $FV$?

### Jawaban & Penjelasan

Pada **Soal 8**, kita diminta menghitung nilai dari anuitas 12 tahun ($R = \$2{,}000$, $i = 6\%$) pada titik waktu perantara **$t = 5$**.

---

#### 1. Mengapa Verifikasinya Menggunakan Rumus $\text{Nilai}_{t=5} = PV_0 \cdot (1+i)^5$?

Terdapat dua cara pandang berbeda untuk menghitung nilai seluruh arus kas pada titik $t = 5$:

##### Cara 1: Dekomposisi Parsial (Membagi Masa Lalu dan Masa Depan)
Titik $t = 5$ membelah 12 pembayaran menjadi 2 kelompok:
1. **5 pembayaran di masa lalu ($t = 1$ s.d. $5$):** Dibawa **maju** ke $t=5$ menggunakan rumus akumulasi $\rightarrow R \cdot s_{\overline{5}\vert 0.06} = \$11{,}274.18$.
2. **7 pembayaran di masa depan ($t = 6$ s.d. $12$):** Dibawa **mundur** ke $t=5$ menggunakan rumus diskonto $\rightarrow R \cdot a_{\overline{7}\vert 0.06} = \$11{,}164.76$.
$$\text{Nilai}_{t=5} = R \cdot s_{\overline{5}\vert} + R \cdot a_{\overline{7}\vert} = \$11{,}274.18 + \$11{,}164.76 = \mathbf{\$22{,}438.94}$$

##### Cara 2: Akumulasi Nilai Sekarang Total ($PV_0 \to t=5$)
Nilai sekarang di $t=0$ ($PV_0 = R \cdot a_{\overline{12}\vert} = \$16{,}767.68$) sudah merangkum seluruh 12 pembayaran menjadi satu nilai tunggal di $t=0$.

$$\text{Nilai}_{t=5} = PV_0 \cdot (1+i)^5 = \$16{,}767.68 \times (1.06)^5 = \mathbf{\$22{,}438.94}$$

##### Bukti Aljabar Identitas Aktuarial:
Kedua cara di atas terbukti identik secara matematis:
$$\begin{aligned}
PV_0 \cdot (1+i)^5 &= \left( R \cdot \frac{1 - v^{12}}{i} \right) (1+i)^5 = R \cdot \left[ \frac{(1+i)^5 - v^7}{i} \right] \\[0.4em]
&= R \cdot \left[ \frac{(1+i)^5 - 1}{i} + \frac{1 - v^7}{i} \right] \\[0.4em]
&= \mathbf{R \cdot s_{\overline{5}\vert} + R \cdot a_{\overline{7}\vert}}
\end{aligned}$$

---

#### 2. Apakah Nilai *Future Value* ($FV$) dan *Present Value* ($PV$) Sama?

Jawabannya tergantung sudut pandang:

1. **Secara Angka Nominal: BERBEDA (Tidak Sama)**

   - Nilai di $t = 5$ (Titik Perantara): **$\$22{,}438.94$**

   Nominalnya terus membesar seiring berjalannya waktu karena adanya bunga majemuk $6\%$ per tahun.

2. **Secara Nilai Ekonomi (*Time Value of Money*): 100% EKUIVALEN (SETARA)**
   - Uang $\$16{,}767.68$ pada saat $t=0$ memiliki daya beli dan nilai finansial yang **persis sama** dengan uang $\$22{,}438.94$ pada $t=5$, dan **persis sama** dengan $\$33{,}739.84$ pada $t=12$.
   - Ketiganya adalah representasi dari **12 arus kas yang sama**, hanya saja dilihat dari "kacamata waktu" yang berbeda di sepanjang garis waktu (*timeline*).

## 12. Mengapa Melangkah Maju 5 Tahun Cukup Dikalikan dengan $(1+i)^5$?

### Pertanyaan
> *"Kenapa maju selama 5 tahun tinggal mengalikannya dengan (1+i)^5?"*

### Jawaban & Penjelasan

Ada dua alasan mendasar: **mekanisme bunga berbunga tahun-demi-tahun** dan **sifat distributif nilai sekarang dari anuitas**.

---

#### 1. Mekanisme Pertumbuhan Bunga Berbunga (*Compound Interest* Tahun demi Tahun)

Misalkan Anda memiliki modal awal sebesar $P$ di titik $t = 0$ dengan suku bunga majemuk $i$ per tahun:

- **Maju 1 Tahun ($t = 0 \to t = 1$):**
  Modal $P$ ditambah bunga sebesar $P \cdot i$:
  $$\text{Saldo}_1 = P + P \cdot i = \mathbf{P \cdot (1+i)}$$

- **Maju 1 Tahun Lagi ($t = 1 \to t = 2$):**
  Seluruh saldo akhir tahun ke-1 dijadikan modal baru yang menghasilkan bunga di tahun ke-2:
  $$\text{Saldo}_2 = \text{Saldo}_1 \cdot (1+i) = [P(1+i)] \cdot (1+i) = \mathbf{P \cdot (1+i)^2}$$

- **Maju 1 Tahun Lagi ($t = 2 \to t = 3$):**
  $$\text{Saldo}_3 = \text{Saldo}_2 \cdot (1+i) = [P(1+i)^2] \cdot (1+i) = \mathbf{P \cdot (1+i)^3}$$

- **Maju 1 Tahun Lagi ($t = 3 \to t = 4$):**
  $$\text{Saldo}_4 = \text{Saldo}_3 \cdot (1+i) = \mathbf{P \cdot (1+i)^4}$$

- **Maju 1 Tahun Lagi ($t = 4 \to t = 5$):**
  $$\text{Saldo}_5 = \text{Saldo}_4 \cdot (1+i) = \mathbf{P \cdot (1+i)^5}$$

> **Kesimpulan:** Setiap kali kita melangkah **maju 1 tahun**, nilai uang dikalikan dengan **$(1+i)$**. Melangkah maju sebanyak **5 tahun berturut-turut** berarti mengalikan $(1+i)$ sebanyak 5 kali:
> $$\underbrace{(1+i) \times (1+i) \times (1+i) \times (1+i) \times (1+i)}_{5 \text{ kali}} = \mathbf{(1+i)^5}$$

---

#### 2. Mengapa Rumus Ini Juga Berlaku untuk Paket Anuitas ($PV_0$)?

Nilai sekarang anuitas $PV_0$ pada dasarnya adalah **gabungan (penjumlahan) dari nilai tunai seluruh pembayaran**:
$$PV_0 = R \cdot v^1 + R \cdot v^2 + R \cdot v^3 + \dots + R \cdot v^{12}$$

Jika kita ingin memindahkan seluruh paket arus kas ini ke titik waktu $t = 5$, secara konsep kita harus memajukan setiap pembayaran sebesar $5$ periode waktu:

$$\begin{aligned}
\text{Arus Kas ke-}1 (t=1) &\text{ di } t=5 \implies R \cdot v^1 \times (1+i)^5 = R(1+i)^4 \\[0.2em]
\text{Arus Kas ke-}2 (t=2) &\text{ di } t=5 \implies R \cdot v^2 \times (1+i)^5 = R(1+i)^3 \\[0.2em]
&\ \ \vdots \\[0.2em]
\text{Arus Kas ke-}12 (t=12) &\text{ di } t=5 \implies R \cdot v^{12} \times (1+i)^5 = R \cdot v^7
\end{aligned}$$

Dengan menggunakan **sifat distributif perkalian** pada aljabar:
$$\begin{aligned}
\text{Total Nilai di } t=5 &= \left( R \cdot v^1 \times (1+i)^5 \right) + \left( R \cdot v^2 \times (1+i)^5 \right) + \dots + \left( R \cdot v^{12} \times (1+i)^5 \right) \\[0.4em]
&= \underbrace{\left( R \cdot v^1 + R \cdot v^2 + \dots + R \cdot v^{12} \right)}_{PV_0} \times \mathbf{(1+i)^5} \\[0.4em]
&= \mathbf{PV_0 \cdot (1+i)^5}
\end{aligned}$$

> **Inti Logika:** $PV_0$ sudah merangkum seluruh 12 arus kas menjadi **satu nilai modal tunggal** di $t=0$. Oleh karena itu, kita tidak perlu memindahkan 12 arus kas satu per satu; cukup pindahkan nilai modal tunggal $PV_0$ tersebut maju 5 tahun dengan mengalikannya dengan **$(1+i)^5$**.

   Jika hari ini di $t = 5$ Anda menggabungkan **saldo tabungan yang sudah ada** ($\$11{,}274.18$) ditambah **setoran tunai untuk melunasi seluruh sisa komitmen masa depan** ($\$11{,}164.76$), total saldo yang langsung terbentuk di rekening Anda adalah **$\$22{,}438.94$**.
   > *(Menariknya: Jika uang $\$22{,}438.94$ ini didiamkan saja berbunga $6\%$ selama 7 tahun tanpa setoran tambahan apa pun lagi, di $t = 12$ saldo akhirnya akan bernilai tepat $\$33{,}739.84$, persis sama dengan hasil menabung rutin 12 tahun!).*

##### Skenario 2: Menjual Kontrak Finansial di Pasar Sekunder (*Secondary Market / Contract Buyout*)

- **Berapa harga jual wajar (*fair market price*) yang harus dibayarkan investor kepada Anda di $t = 5$?**
  Investor akan menghitung nilai tunai (*Present Value*) dari 7 penerimaan masa depan tersebut:
  $$\text{Harga Jual Sisa Kontrak} = 2{,}000 \cdot a_{\overline{7}\vert 0.06} = \mathbf{\$11{,}164.76}$$
- **Makna Nilai Total $\$22{,}438.94$:**
  Total seluruh kekayaan yang Anda peroleh dari kontrak 12 tahun tersebut pada titik $t = 5$ adalah uang tunai hasil 5 tahun pertama ($\$11{,}274.18$) ditambah uang hasil penjualan sisa kontrak hari ini ($\$11{,}164.76$), yaitu **$\$22{,}438.94$**.

##### Skenario 3: Pelunasan dan Restrukturisasi Utang KPR / Pinjaman Bisnis (*Loan Balance & Refinancing*)
Bayangkan Anda meminjam modal di bank pada $t = 0$ sebesar **$PV_0 = \$16{,}767.68$** dengan suku bunga $6\%$ per tahun, yang disepakati untuk dilunasi dengan cicilan **$\$2{,}000$ per tahun selama 12 tahun**.

Pada akhir tahun ke-5 ($t = 5$):
1. **Jika Anda tidak pernah membayar cicilan sama sekali selama 5 tahun**, total utang pokok Anda beserta bunganya akan membengkak menjadi:
   $$\text{Utang Kotor Terakumulasi} = \$16{,}767.68 \times (1.06)^5 = \mathbf{\$22{,}438.94}$$

   $$\text{Sisa Utang Pelunasan} = \underbrace{\$22{,}438.94}_{\text{Utang Kotor}} - \underbrace{\$11{,}274.18}_{\text{Cicilan Terbayar}} = \mathbf{\$11{,}164.76}$$

4. **Makna Nilai $\$22{,}438.94$:**
   Ini adalah **tolok ukur utang kotor (*gross debt accumulated*)** atau nilai waktu dari pokok pinjaman awal $PV_0$ yang telah berjalan selama 5 tahun sebelum dikompensasikan dengan pembayaran yang telah dilakukan.

## 14. Latihan Soal 05 (Soal 10): Mengapa Masalah Valuasi Mesin Industri & NPV Bisa Dikerjakan dengan Konsep Anuitas?

### Pertanyaan
> *"Saya tidak paham kenapa soal 10 bisa dikerjakan dengan konsep anuitas."*

### Jawaban & Penjelasan

Banyak orang mengira bahwa **anuitas** hanyalah istilah untuk *"cicilan kredit di bank"* atau *"uang pensiun bulanan"*. Padahal dalam matematika aktuaria dan keuangan, definisi anuitas jauh lebih luas dan universal.

---

#### 1. Definisi Fundamental Anuitas: Syarat Suatu Arus Kas Disebut Anuitas

Suatu rangkaian arus kas (**apapun bentuknya di dunia nyata**) secara matematis disebut sebagai **ANUITAS** jika memenuhi **3 syarat utama**:
1. **Besaran Nominal Tetap / Konstan ($R$):** Setiap kali terjadi transaksi, jumlah uangnya sama.
2. **Interval Waktu Teratur / Periodik:** Jarak waktu antar arus kas selalu sama (misalnya setiap 1 tahun, setiap 1 bulan).
3. **Durasi Waktu Berhingga ($n$):** Berlangsung selama sejumlah periode tertentu.

> **Kesimpulan Kunci:**
> Tidak peduli apakah uang itu berupa **penghematan biaya listrik/tenaga kerja**, **biaya servis pemeliharaan rutin**, sewa gedung, dividen saham, atau cicilan utang—selama nominalnya tetap dan terjadi secara berkala, ia adalah **ANUITAS**!

---

#### 2. Bedah 4 Komponen Arus Kas pada Soal 10

Pada **Soal 10**, sebuah pabrik ingin membeli mesin seharga $\$50{,}000$ dengan biaya modal $i = 8\%$ selama $8$ tahun masa pakai. Proyek ini memicu **4 aliran uang yang berbeda karakteristiknya**:

| Komponen Arus Kas | Pola & Waktu Kejadian | Klasifikasi Matematis | Formula Nilai Sekarang ($PV$) |
| :--- | :--- | :--- | :---: |
| **1. Biaya Pembelian Awal** | Keluar $\$50{,}000$ tepat di **hari ini ($t = 0$)** | **Kas Tunggal (*Lump Sum*)** di $t = 0$ | $-\$50{,}000$ |
| **2. Penghematan Tenaga Kerja** | Hemat $\$9{,}000$ di **setiap akhir tahun** ($t = 1, 2, \dots, 8$) | **Anuitas Akhir (*Annuity-Immediate*)** | $\mathbf{+9{,}000 \cdot a_{\overline{8}\vert 0.08}}$ |
| **3. Biaya Servis Rutin** | Bayar $\$1{,}200$ di **setiap awal tahun** ($t = 0, 1, \dots, 7$) | **Anuitas Awal (*Annuity-Due*)** | $\mathbf{-1{,}200 \cdot \ddot{a}_{\overline{8}\vert 0.08}}$ |
| **4. Nilai Sisa (*Salvage Value*)** | Dijual rongsok $\$5{,}000$ di **akhir tahun ke-8 ($t = 8$)** | **Kas Tunggal (*Lump Sum*)** di $t = 8$ | $+5{,}000 \cdot (1.08)^{-8}$ |

---

#### 3. Mengapa Penghematan dan Servis Merupakan Anuitas?

##### A. Penghematan Biaya Tenaga Kerja ($+\$9{,}000$ per tahun)
- Mesin menghemat pengeluaran gaji karyawan sebesar $\$9{,}000$ setiap akhir tahun selama 8 tahun ($t = 1, 2, \dots, 8$).
- Dari sudut pandang pabrik, uang $\$9{,}000$ yang tidak jadi dibayarkan kepada pekerja adalah **arus kas masuk bersih (*cash inflow*)**.
- Karena diterima di **akhir periode** secara rutin sebanyak 8 kali, ini persis memenuhi definisi **Anuitas Akhir (*Annuity-Immediate*)**:
  $$PV_{\text{hemat}} = 9{,}000 \cdot v^1 + 9{,}000 \cdot v^2 + \dots + 9{,}000 \cdot v^8 = \mathbf{9{,}000 \cdot a_{\overline{8}\vert 0.08} = +\$51{,}719.75}$$

##### B. Biaya Servis Tahunan ($-\$1{,}200$ per tahun)
- Kontrak servis mesin mewajibkan pembayaran di **setiap awal tahun** ($t = 0, 1, 2, \dots, 7$, total 8 kali bayar).
- Karena pembayaran dilakukan di **muka / awal periode**, ini persis memenuhi definisi **Anuitas Awal (*Annuity-Due*)**:
  $$PV_{\text{servis}} = 1{,}200 + 1{,}200 \cdot v^1 + \dots + 1{,}200 \cdot v^7 = \mathbf{1{,}200 \cdot \ddot{a}_{\overline{8}\vert 0.08} = -\$7{,}447.64}$$

#### 5. Keuntungan Menggunakan Rumus Anuitas: Mengapa Tidak Dihitung Manual Satu per Satu?

Bayangkan jika Anda menghitung nilai sekarang dari setiap transaksi satu per satu tanpa konsep anuitas:
$$\begin{aligned}
NPV &= -50{,}000 + \sum_{t=1}^8 \frac{9{,}000}{(1.08)^t} - \sum_{t=0}^7 \frac{1{,}200}{(1.08)^t} + \frac{5{,}000}{(1.08)^8} \\[0.4em]
&= -50{,}000 + \underbrace{\left[ \frac{9000}{1.08^1} + \dots + \frac{9000}{1.08^8} \right]}_{8 \text{ suku penghematan}} - \underbrace{\left[ 1200 + \frac{1200}{1.08^1} + \dots + \frac{1200}{1.08^7} \right]}_{8 \text{ suku biaya servis}} + \frac{5000}{1.08^8}
\end{aligned}$$

Anda harus menghitung **18 operasi pemangkatan dan pembagian pecahan desimal**!

Dengan konsep **Anuitas**, 16 suku berkala tersebut diringkas menjadi hanya **2 simbol terfaktor**:
$$\mathbf{NPV = -50{,}000 + 9{,}000 \cdot a_{\overline{8}\vert 0.08} - 1{,}200 \cdot \ddot{a}_{\overline{8}\vert 0.08} + 5{,}000 \cdot v^8}$$

$$\begin{aligned}
NPV &= -50{,}000 + (9{,}000 \times 5.746639) - (1{,}200 \times 6.206370) + (5{,}000 \times 0.540269) \\[0.3em]
&= -50{,}000 + 51{,}719.75 - 7{,}447.64 + 2{,}701.35 = \mathbf{-\$3{,}026.54}
\end{aligned}$$

#### 6. Kesimpulan & Relevansi Dunia Nyata (*Capital Budgeting*)

Dalam dunia nyata (analisis investasi bisnis, perbankan investasi, dan aktuaria perusahaan):
1. Hampir semua proyek industri (pembelian mesin pabrik, ekspansi armada logistik, instalasi panel surya) melibatkan **pengeluaran rutin** (biaya *maintenance*) dan **penerimaan rutin** (penjualan / penghematan operasional).
2. Konsep anuitas adalah **alat tercepat dan paling elegan** untuk mengonversi seluruh aliran biaya dan manfaat masa depan tersebut menjadi **satu nilai tunai di hari ini ($NPV$)**.
3. Karena $NPV = -\$3{,}026.54 < 0$, artinya jika pabrik membeli mesin tersebut, keuntungan yang dihasilkan kalah dibandingkan jika uang $\$50{,}000$ diinvestasikan pada instrumen modal lain yang menghasilkan $8\%$. Oleh sebab itu, investasi **ditolak**.

```python
# Verifikasi Numerik Lengkap: Soal 8 & Soal 10 (Analisis Arus Kas Anuitas)

# ==========================================================================
# SOAL 8: Evaluasi Titik Waktu Perantara t = 5 (Dekomposisi vs PV_0 Akumulasi)
# ==========================================================================
i_8 = 0.06
R_8 = 2000.0
n_8 = 12
t_split = 5
t_rem = n_8 - t_split  # 7

s_5 = ((1 + i_8)**t_split - 1) / i_8
a_7 = (1 - (1 + i_8)**(-t_rem)) / i_8
past_val = R_8 * s_5       # $11,274.18
future_val = R_8 * a_7     # $11,164.76
val_split = past_val + future_val

a_12 = (1 - (1 + i_8)**(-n_8)) / i_8
pv_0_8 = R_8 * a_12
val_accum = pv_0_8 * ((1 + i_8)**t_split)

print("="*75)
print("SOAL 8: VALUASI ANUITAS PADA TITIK PERANTARA t = 5")
print("="*75)
print(f"- Nilai Sekarang di t = 0 (PV_0)                   : ${pv_0_8:,.2f}")
print(f"- Saldo riil tabungan masa lalu di bank (s_5)     : ${past_val:,.2f}")
print(f"- Nilai tunai sisa 7 komitmen masa depan (a_7)    : ${future_val:,.2f}")
print(f"- Total Nilai di t = 5 (Dekomposisi s_5 + a_7)     : ${val_split:,.2f}")
print(f"- Total Nilai di t = 5 (Akumulasi PV_0 * (1+i)^5)  : ${val_accum:,.2f} (Identik!)")

# ==========================================================================
# SOAL 10: Valuasi Mesin Industri (NPV dengan Formula Anuitas vs Manual)
# ==========================================================================
i_10 = 0.08
cost_initial = 50000.0
R_hemat = 9000.0      # Annuity-immediate (t = 1..8)
R_servis = 1200.0     # Annuity-due (t = 0..7)
salvage = 5000.0       # Lump sum di t = 8
n_years = 8

# 1. Menggunakan Rumus Anuitas (Cara Cepat Aktuaria)
a_8_008 = (1 - (1 + i_10)**(-n_years)) / i_10
a_due_8_008 = (1 + i_10) * a_8_008
v_8 = (1 + i_10)**(-n_years)

pv_hemat_annuity = R_hemat * a_8_008
pv_servis_annuity = R_servis * a_due_8_008
pv_salvage = salvage * v_8
npv_annuity = -cost_initial + pv_hemat_annuity - pv_servis_annuity + pv_salvage

# 2. Menggunakan Perhitungan Arus Kas Manual per Tahun
net_cf_per_year = [-cost_initial - R_servis]  # t = 0
for t in range(1, 8):
    net_cf_per_year.append(R_hemat - R_servis)  # t = 1..7 (9000 - 1200 = 7800)
net_cf_per_year.append(R_hemat + salvage)       # t = 8 (9000 + 5000 = 14000)

npv_manual = sum(cf / ((1 + i_10)**t) for t, cf in enumerate(net_cf_per_year))

print("="*75)
print("SOAL 10: ANALISIS KELAYAKAN INVESTASI MESIN (NPV & ANUITAS)")
print("="*75)
print(f"1. Biaya Pembelian Mesin di t = 0                  : -${cost_initial:,.2f}")
print(f"2. PV Penghematan Tenaga Kerja (Anuitas Akhir)     : +${pv_hemat_annuity:,.2f}")
print(f"3. PV Biaya Servis Rutin (Anuitas Awal)            : -${pv_servis_annuity:,.2f}")
print(f"4. PV Nilai Sisa Mesin di t = 8 (Lump Sum)         : +${pv_salvage:,.2f}")
print("-"*75)
print(f"- Net Present Value (Rumus Anuitas)                : ${npv_annuity:,.2f}")
print(f"- Net Present Value (Hitung Manual Arus Kas)       : ${npv_manual:,.2f} (100% Cocok!)")
print(f"- Kesimpulan Keputusan Investasi                   : {'LAYAK (DITERIMA)' if npv_annuity >= 0 else 'TIDAK LAYAK (DITOLAK)'}")
print("="*75)
```
