### B. Skenario Konkret di Dunia Nyata: "Tabungan Rencana dengan Kenaikan Gaji"

Bayangkan seseorang, sebut saja **Budi**, yang membuka rekening **Tabungan Rencana / Reksa Dana Otomatis** selama 4 tahun dengan skenario sebagai berikut:

#### 2. Realitas Sistem Perbankan (*Mismatch* Frekuensi Bunga dan Setoran)
- Bank atau manajer investasi tempat Budi menabung menghitung bunga **setiap akhir bulan** dengan suku bunga nominal tahunan $12\%$.
- Artinya, suku bunga efektif bulanan yang dikreditkan ke saldo rekening adalah:
  $$i_{\text{bulan}} = \frac{12\%}{12} = 1\% = 0.01 \text{ per bulan}$$
- Di sinilah letak alasan materi ini disebut **Anuitas yang Lebih Umum (*General Annuities*)**:

#### 3. Apa Arti "Nilai Akumulasi pada Akhir Tahun ke-4" ($AV_{16}$)?

2. Seluruh hasil bunga majemuk bergulir (*compound interest*) yang terus berputar selama 4 tahun.

Hasil perhitungannya adalah **$\$2{,}998.86$**.
Artinya:
- Modal disetor: $\$2{,}400.00$
- Bunga bersih yang diperoleh: $\$2{,}998.86 - \$2{,}400.00 = \mathbf{\$598.86}$

### C. Garis Waktu Arus Kas (*Cash Flow Timeline*)

>- Karena ada selisih tepat 1 kuartal antara setoran terakhir dengan titik evaluasi, formula yang digunakan adalah **Anuitas Awal ($\ddot{s}$)**.

### D. Penurunan Matematis Langkah demi Langkah

#### Langkah 1: Suku Bunga Efektif per Kuartal ($j$)
Karena bunga dibungakan bulanan dengan $i_{\text{bulan}} = 1\% = 0.01$, dalam 1 kuartal (3 bulan) uang bertumbuh sebesar:
$$1 + j = (1 + i_{\text{bulan}})^3 = (1 + 0.01)^3 = 1.030301$$
$$j = 0.030301 \quad (3.0301\% \text{ per kuartal})$$

Tingkat diskon efektif per kuartal ($d_j$) adalah:
$$d_j = \frac{j}{1+j} = \frac{0.030301}{1.030301} \approx 0.0294099$$

---

#### Langkah 2: Dua Cara Pandang Menyelesaikan Arus Kas

##### Cara 1: Metode Pelapisan Anuitas (*Layering Method*)
Kita dapat memandang tabungan Budi sebagai dua lapis anuitas:
1. **Lapisan Dasar (Layer 1):** Setoran $\$100$ yang berjalan penuh sepanjang 16 kuartal ($t=0$ s.d. $15$).
2. **Lapisan Tambahan (Layer 2):** Tambahan setoran $\$100$ (karena $\$200 - \$100 = \$100$) yang berjalan selama 8 kuartal terakhir ($t=8$ s.d. $15$).

$$\begin{aligned}
AV_{16} &= 100 \cdot \ddot{s}_{\overline{16}\vert j} + 100 \cdot \ddot{s}_{\overline{8}\vert j} \\[0.5em]
&= 100 \left[ \frac{(1.030301)^{16} - 1}{0.0294099} \right] + 100 \left[ \frac{(1.030301)^8 - 1}{0.0294099} \right] \\[0.5em]
&= 100 \left[ \frac{1.6122261 - 1}{0.0294099} \right] + 100 \left[ \frac{1.2697346 - 1}{0.0294099} \right] \\[0.5em]
&= 100 (20.81704) + 100 (9.17157) \\[0.5em]
&= \$2{,}081.70 + \$917.16 = \mathbf{\$2{,}998.86}
\end{aligned}$$

---

##### Cara 2: Metode Kronologis Berurutan (*Sequential Accumulation*)
Kita juga bisa menghitung saldo rekening sesuai urutan waktu berjalannya hidup Budi:

1. **Hitung Saldo di Akhir Tahun ke-2 ($t = 8$):**
   Akumulasi 8 setoran pertama sebesar $\$100$:
   $$AV_8 = 100 \cdot \ddot{s}_{\overline{8}\vert j} = 100 \times 9.17157 = \$917.16$$

2. **Akumulasikan Saldo Tersebut 2 Tahun ke Depan ($t = 8 \to t = 16$):**
   Saldo $\$917.16$ tersebut dibiarkan terus berbunga selama 8 kuartal berikutnya:
   $$\text{Nilai Pertumbuhan Saldo Lama} = \$917.16 \times (1+j)^8 = \$917.16 \times 1.269735 = \mathbf{\$1{,}164.55}$$

3. **Hitung Akumulasi dari 8 Setoran Baru ($\$200$) di Tahun 3 & 4:**
   $$\text{Nilai Setoran Baru} = 200 \cdot \ddot{s}_{\overline{8}\vert j} = 200 \times 9.17157 = \mathbf{\$1{,}834.31}$$

4. **Total Saldo Akhir di $t = 16$:**
   $$AV_{16} = \$1{,}164.55 + \$1{,}834.31 = \mathbf{\$2{,}998.86}$$

> Kedua metode menghasilkan angka yang **100% identik**, membuktikan konsistensi hukum nilai waktu dari uang (*Time Value of Money*).

```python
# Verifikasi Komputasi dengan Python
# Simulasi Perjalanan Saldo Rekening Budi Tiap Kuartal

i_month = 0.12 / 12  # 1% per bulan
j = (1 + i_month)**3 - 1  # Suku bunga efektif per kuartal
d_j = j / (1 + j)         # Tingkat diskon efektif per kuartal

print(f"Suku bunga efektif per kuartal (j) : {j:.6%}")
print(f"Tingkat diskon per kuartal (d_j)   : {d_j:.6%}")

# Simulasi Mutasi Rekening Buku Tabungan
balance = 0.0
print("\n--- BUKU TABUNGAN REKENING BUDI (PER KUARTAL) ---")
print(f"{'Kuartal':<8}{'Setoran':<12}{'Saldo Awal':<15}{'Bunga Kuartal':<15}{'Saldo Akhir':<15}")

for q in range(16):
    deposit = 100.0 if q < 8 else 200.0
    start_balance = balance + deposit
    interest = start_balance * j
    balance = start_balance + interest
    print(f"{q+1:<8}{deposit:<12.2f}{start_balance:<15.2f}{interest:<15.2f}{balance:<15.2f}")

print("-" * 65)
print(f"TOTAL SALDO AKHIR DI AKHIR TAHUN KE-4: ${balance:.2f}")
```

## 2. Cara Menentukan Suku Bunga Efektif ($j$) dari Bunga Nominal: Mengapa Letak Pangkat Berbeda pada Contoh 1 dan Contoh 2?

### Pertanyaan
> *"Saya kurang paham bagaimana cara menuliskan suku bunga efektif jika diketahui bunga nominalnya. Saya liat pada contoh soal 1 dan contoh soal 2 berbeda mana yang di pangkatkan"*

---

### A. Prinsip Dasar: Apa Target yang Ingin Dicari?

Ketika menyelesaikan soal anuitas umum dengan Pendekatan Numerik:
1. **Target ($j$):** Kita selalu ingin mencari suku bunga efektif per **periode pembayaran transaksi** (misalnya per bulan, per kuartal, per semester, dll).
2. **Modal Dasar ($i_{\text{dasar}}$):** Bunga nominal $i^{(m)}$ selalu dibagi dengan frekuensi konversinya ($m$) untuk mendapatkan bunga efektif per periode pemajemukan bank:
   $$i_{\text{dasar}} = \frac{i^{(m)}}{m}$$

> **Pertanyaan Kuncinya:**
> *"Bagaimana menghubungkan $j$ (periode pembayaran) dengan $i_{\text{dasar}}$ (periode bunga bank)?"*

---

### B. Mengapa Pangkatnya Berbeda? (Perbandingan Panjang Durasi)

Perhatikan perbandingan durasi waktu pada kedua contoh soal:

#### 1. Kasus Contoh 1: Periode Pembayaran LEBIH PANJANG daripada Periode Bunga
- **Periode Pembayaran Transaksi:** Kuartalan (**3 bulan**).
- **Periode Pemajemukan Bank:** Bulanan (**1 bulan**), dengan $i_{\text{bulan}} = \frac{12\%}{12} = 1\% = 0.01$.

**Logika Pertumbuhan Uang:**
- Satu periode pembayaran (3 bulan) menampung **3 kali putaran bunga bulanan**.
- Jadi, pertumbuhan uang selama 1 kuartal ($1+j$) adalah hasil perkalian bunga bulanan sebanyak 3 kali berturut-turut:
  $$\mathbf{1 + j = (1 + i_{\text{bulan}})^3}$$
  $$1 + j = (1 + 0.01)^3 = 1.030301 \implies \mathbf{j = (1.01)^3 - 1 = 0.030301}$$
- 👉 *Pangkat $3$ (bilangan bulat $> 1$) berada di sisi suku bunga bulanan $(1+i_{\text{bulan}})$.*

---

#### 2. Kasus Contoh 2: Periode Pembayaran LEBIH PENDEK daripada Periode Bunga
- **Periode Pembayaran Transaksi:** Kuartalan (**3 bulan**).
- **Periode Pemajemukan Bank:** Semesteran (**6 bulan**), dengan $i_{\text{semester}} = \frac{10\%}{2} = 5\% = 0.05$.

**Logika Pertumbuhan Uang:**
- Satu periode bunga bank (6 bulan) menampung **2 periode pembayaran kuartal**.
- Jadi, jika uang bertumbuh dengan bunga kuartalan $j$ sebanyak 2 kali berturut-turut, hasilnya harus tepat sama dengan pertumbuhan 1 semester:
  $$\mathbf{(1 + j)^2 = 1 + i_{\text{semester}}}$$
  $$(1 + j)^2 = 1 + 0.05 = 1.05$$
- Untuk mencari $(1+j)$, kita tarik akar (pangkatkan kedua ruas dengan $\frac{1}{2}$):
  $$\mathbf{1 + j = (1.05)^{1/2} = \sqrt{1.05}} \implies \mathbf{j = (1.05)^{1/2} - 1 \approx 0.024695}$$
- 👉 *Pangkat $2$ berada di sisi $(1+j)$, yang ketika dipindahkan ke ruas kanan berubah menjadi pangkat pecahan $\frac{1}{2}$.*

### C. Metode Universal Anti-Bingung: Formula Penyetaraan 1 Tahun (*Golden Rule*)

Jika Anda tidak ingin menghafal kasus mana yang lebih panjang atau pendek, gunakan **Satu Prinsip Emas Universal** ini:

> **"Nilai akumulasi dari modal $\$1$ setelah mengendap selama 1 TAHUN PENUH harus sama persis di kedua sudut pandang."**

Definisikan:
- **$p$** = Frekuensi pembayaran transaksi dalam 1 tahun ($p = 4$ untuk kuartal, $p = 12$ untuk bulan, $p = 2$ untuk semester, $p = 1$ untuk tahunan).
- **$m$** = Frekuensi pemajemukan bunga nominal bank dalam 1 tahun ($m$ dari simbol $i^{(m)}$).

Maka dalam rentang waktu **1 tahun**:
$$\underbrace{(1 + j)^p}_{\text{Akumulasi setahun dari } p \text{ periode bayar}} = \underbrace{\left( 1 + \frac{i^{(m)}}{m} \right)^m}_{\text{Akumulasi setahun dari } m \text{ periode konversi bank}}$$

Tarik akar ke-$p$ (pangkatkan kedua ruas dengan $\frac{1}{p}$):

$$\mathbf{1 + j = \left( 1 + \frac{i^{(m)}}{m} \right)^{\frac{m}{p}}}$$

$$\mathbf{j = \left( 1 + \frac{i^{(m)}}{m} \right)^{\frac{m}{p}} - 1}$$

---

### D. Uji Coba Rumus Universal pada Kedua Contoh Soal

Mari kita buktikan betapa mudahnya menggunakan formula universal $\mathbf{\frac{m}{p}}$:

| Komponen | Contoh Soal 1 | Contoh Soal 2 |
| :--- | :--- | :--- |
| **Frekuensi Bayar ($p$)** | Kuartalan $\implies \mathbf{p = 4}$ | Kuartalan $\implies \mathbf{p = 4}$ |
| **Frekuensi Bunga ($m$)** | Bulanan $\implies \mathbf{m = 12}$ | Semesteran $\implies \mathbf{m = 2}$ |
| **Bunga Dasar $\left(\frac{i^{(m)}}{m}\right)$** | $\frac{12\%}{12} = 1\% = 0.01$ | $\frac{10\%}{2} = 5\% = 0.05$ |
| **Pangkat Rasio $\left(\frac{m}{p}\right)$** | $\dfrac{m}{p} = \dfrac{12}{4} = \mathbf{3}$ | $\dfrac{m}{p} = \dfrac{2}{4} = \mathbf{\dfrac{1}{2} = 0.5}$ |
| **Persamaan $j$** | $j = (1 + 0.01)^{\mathbf{3}} - 1$ | $j = (1 + 0.05)^{\mathbf{1/2}} - 1$ |
| **Hasil Akhir $j$** | $\mathbf{3.030101\%}$ per kuartal | $\mathbf{2.469508\%}$ per kuartal |

> **Ringkasan Praktis:**
>- Jika $\frac{m}{p} > 1$ (seperti Contoh 1: $\frac{12}{4} = 3$), pangkatnya berupa bilangan bulat $> 1$.
>- Jika $\frac{m}{p} < 1$ (seperti Contoh 2: $\frac{2}{4} = 0.5$), pangkatnya berupa pecahan (akar).

```python
# Kalkulator Universal Suku Bunga Efektif per Periode Pembayaran (j)

def hitung_suku_bunga_efektif(i_nominal, m, p):
    """
    Menghitung suku bunga efektif per periode pembayaran (j)
    Parameter:
      i_nominal : Suku bunga nominal tahunan (misal 0.12 untuk 12%)
      m         : Frekuensi pemajemukan bunga per tahun (misal 12 untuk bulanan, 2 untuk semesteran)
      p         : Frekuensi pembayaran transaksi per tahun (misal 4 untuk kuartalan, 12 untuk bulanan)
    """
    i_dasar = i_nominal / m
    rasio_pangkat = m / p
    j = (1 + i_dasar)**rasio_pangkat - 1
    return i_dasar, rasio_pangkat, j

# Uji Kasus Contoh 1: i^(12) = 12%, bayar kuartalan (p=4)
i_dasar1, rasio1, j1 = hitung_suku_bunga_efektif(0.12, m=12, p=4)
print("=== CONTOH SOAL 1 ===")
print(f"Bunga per periode bank (bulanan) : {i_dasar1:.4%}")
print(f"Pangkat rasio (m/p = 12/4)       : {rasio1:.1f}")
print(f"Suku bunga efektif per kuartal (j): {j1:.6%}")

# Uji Kasus Contoh 2: i^(2) = 10%, bayar kuartalan (p=4)
i_dasar2, rasio2, j2 = hitung_suku_bunga_efektif(0.10, m=2, p=4)
print("\n=== CONTOH SOAL 2 ===")
print(f"Bunga per periode bank (semester): {i_dasar2:.4%}")
print(f"Pangkat rasio (m/p = 2/4)        : {rasio2:.2f}")
print(f"Suku bunga efektif per kuartal (j): {j2:.6%}")
```

## 3. Interpretasi Rumus Pokok: Pembayaran 1 Setiap Awal $k$ Periode Bunga ($PV = \frac{a_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}$)

### Pertanyaan
> *"Saya tidak paham interpretasi pada 'Rumus Pokok: Pembayaran 1 Setiap Awal k Periode Bunga' (Catatan 07 Subbab 3.3)"*

---

### A. Bunyi Rumus & Pernyataan dalam Catatan 07

Pada Subbab 3.3, untuk transaksi di mana uang sebesar **$\$1$** dibayarkan pada **setiap awal $k$ periode bunga** ($t = 0, k, 2k, \dots, n-k$):
$$\mathbf{PV = \frac{a_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}} \qquad \text{dan} \qquad \mathbf{FV = \frac{s_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}}$$

> **Interpretasi Aktuarial:**
> *"Pembayaran $1$ pada awal setiap interval sepanjang $k$ periode bunga dapat digantikan oleh pembayaran berkala sebesar $\frac{1}{a_{\overline{k}\vert i}}$ pada setiap akhir periode bunga tunggal."*

---

### B. Mengapa Bisa Digantikan? (Intuisi Finansial Konkret)

Mari kita bedah secara logis bagaimana uang $\$1$ di awal interval bisa digantikan dengan cicilan $\frac{1}{a_{\overline{k}\vert i}}$:

#### 1. Tinjau 1 Blok Interval Saja ($t = 0$ sampai $t = k$)
- Bayangkan pada saat **$t = 0$ (hari ini)**, Anda menerima uang tunai sebesar **$\$1$**.
- Menerima uang tunai $\$1$ di awal hari ini secara finansial ekuivalen dengan **mengambil pinjaman sebesar $\$1$** di $t=0$.
- Jika Anda memiliki utang $\$1$ di $t=0$ dan ingin melunasinya dengan mencicil tetap sebesar $C$ di **setiap akhir periode** selama $k$ periode berturut-turut ($t = 1, 2, \dots, k$), berapa besarnya cicilan $C$?
  $$1 = C \cdot a_{\overline{k}\vert i} \implies \mathbf{C = \frac{1}{a_{\overline{k}\vert i}}}$$

> **Kesimpulan Blok 1:**
> Menerima uang tunai **$\$1$ sekaligus di awal ($t=0$)** memiliki nilai ekonomi yang **100% SAMA** dengan menerima **aliran cicilan sebesar $C = \frac{1}{a_{\overline{k}\vert i}}$ di setiap $t = 1, 2, \dots, k$**.

---

#### 2. Terapkan Prinsip Penggantian Ini ke Seluruh Blok Pembayaran

Jika kita ganti setiap setoran $\$1$ gelondongan dengan serangkaian cicilan ekuivalennya:
1. **Setoran $\$1$ di $t=0$** $\longrightarrow$ Diganti cicilan $C = \frac{1}{a_{\overline{k}\vert i}}$ pada $t = 1, 2, \dots, k$.
2. **Setoran $\$1$ di $t=k$** $\longrightarrow$ Diganti cicilan $C = \frac{1}{a_{\overline{k}\vert i}}$ pada $t = k+1, k+2, \dots, 2k$.
3. **Setoran $\$1$ di $t=2k$** $\longrightarrow$ Diganti cicilan $C = \frac{1}{a_{\overline{k}\vert i}}$ pada $t = 2k+1, \dots, 3k$.
4. $\dots$
5. **Setoran $\$1$ di $t=n-k$** $\longrightarrow$ Diganti cicilan $C = \frac{1}{a_{\overline{k}\vert i}}$ pada $t = n-k+1, \dots, n$.

---

#### 3. Garis Waktu Baru yang Tercipta

Perhatikan apa yang terjadi pada seluruh garis waktu dari $t = 1$ hingga $t = n$:
- Pembayaran gelondongan yang jarang-jarang (setiap $k$ periode) kini telah **lebur/terdistribusi merata** menjadi **pembayaran konstan sebesar $C = \frac{1}{a_{\overline{k}\vert i}}$ pada SETIAP akhir periode bunga tanpa jeda** dari $t = 1$ sampai $t = n$.
- Rangkaian pembayaran baru ini tidak lain adalah **Anuitas Akhir Biasa (*Standard Annuity-Immediate*)** berjangka $n$ periode dengan besar pembayaran $C$!

Maka Nilai Sekarang ($PV$) dari seluruh transaksi adalah:
$$PV = C \cdot a_{\overline{n}\vert i} = \left( \frac{1}{a_{\overline{k}\vert i}} \right) \cdot a_{\overline{n}\vert i} = \mathbf{\frac{a_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}}$$

Dan Nilai Terakumulasi ($FV$) pada $t = n$ adalah:
$$FV = C \cdot s_{\overline{n}\vert i} = \left( \frac{1}{a_{\overline{k}\vert i}} \right) \cdot s_{\overline{n}\vert i} = \mathbf{\frac{s_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}}$$

> Inilah rahasia mengapa simbol $a_{\overline{k}\vert i}$ muncul sebagai penyebut pembagi!

### C. Perbandingan Kontras: Anuitas Akhir vs Anuitas Awal ($k > 1$)

Untuk melihat keindahan simetrisnya, mari bandingkan kedua kasus pembagian ini:

| Karakteristik | Kasus Anuitas Akhir ($k > 1$) | Kasus Anuitas Awal ($k > 1$) |
| :--- | :--- | :--- |
| **Titik Pembayaran** | Di **AKHIR** interval ($t = k, 2k, \dots, n$) | Di **AWAL** interval ($t = 0, k, \dots, n-k$) |
| **Analogi Keuangan** | *Sinking Fund* (menabung $C$ tiap periode untuk mengumpulkan $\$1$ di **akhir** periode ke-$k$) | *Loan Amortization* (mencicil $C$ tiap periode untuk melunasi utang $\$1$ yang diterima di **awal**) |
| **Nilai Cicilan Pengganti ($C$)** | $1 = C \cdot s_{\overline{k}\vert i} \implies \mathbf{C = \dfrac{1}{s_{\overline{k}\vert i}}}$ | $1 = C \cdot a_{\overline{k}\vert i} \implies \mathbf{C = \dfrac{1}{a_{\overline{k}\vert i}}}$ |
| **Pembagi Formula** | Dibagi **$s_{\overline{k}\vert i}$** | Dibagi **$a_{\overline{k}\vert i}$** |
| **Rumus Nilai Sekarang ($PV$)** | $\mathbf{PV = \dfrac{a_{\overline{n}\vert i}}{s_{\overline{k}\vert i}}}$ | $\mathbf{PV = \dfrac{a_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}}$ |
| **Rumus Nilai Akumulasi ($FV$)** | $\mathbf{FV = \dfrac{s_{\overline{n}\vert i}}{s_{\overline{k}\vert i}}}$ | $\mathbf{FV = \dfrac{s_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}}$ |

---

### D. Pembuktian Aljabar Murni Deret Geometri

Jika Anda ingin melihat penurunan matematis murninya tanpa analogi cicilan:
- Arus kas terjadi di $t = 0, k, 2k, \dots, n-k$ sebanyak $m = n/k$ kali pembayaran.
- Diskontokan masing-masing ke $t = 0$:
  $$PV = 1 + v^k + v^{2k} + \dots + v^{n-k}$$
- Jumlah deret geometri dengan rasio $r = v^k$ dan $m$ suku:
  $$PV = \frac{1 - (v^k)^m}{1 - v^k} = \frac{1 - v^n}{1 - v^k}$$
- Bagi pembilang dan penyebut dengan $i$:
  $$PV = \frac{\dfrac{1 - v^n}{i}}{\dfrac{1 - v^k}{i}} = \mathbf{\frac{a_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}}$$

### E. Contoh Angka Nyata

Misalkan suku bunga $i = 10\%$ per tahun. Anda dijanjikan menerima **$\$1{,}000$ di awal setiap 3 tahun ($k = 3$)** selama total jangka waktu **6 tahun ($n = 6$)**.
- Artinya, Anda menerima $\$1{,}000$ di $t = 0$ dan $\$1{,}000$ di $t = 3$.

**Cara 1 (Konsep Penggantian Cicilan):**
Uang $\$1{,}000$ di awal setiap 3 tahun digantikan oleh cicilan tahunan:
$$C = \frac{1000}{a_{\overline{3}\vert 0.10}} = \frac{1000}{2.48685} = \$402.11 \text{ per tahun selama 6 tahun}$$
Maka $PV = 402.11 \times a_{\overline{6}\vert 0.10} = 402.11 \times 4.35526 = \mathbf{\$1{,}751.31}$

**Cara 2 (Rumus Langsung $\frac{a_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}$):**
$$PV = 1000 \cdot \frac{a_{\overline{6}\vert 0.10}}{a_{\overline{3}\vert 0.10}} = 1000 \times \frac{4.35526}{2.48685} = \mathbf{\$1{,}751.31}$$

**Cara 3 (Diskonto Manual Kas Tunggal):**
$$PV = 1000 + 1000(1.10)^{-3} = 1000 + 751.31 = \mathbf{\$1{,}751.31}$$

> Ketiga cara menghasilkan angka yang **100% sama**, membuktikan bahwa interpretasi penggantian cicilan $1/a_{\overline{k}\vert i}$ benar-benar bekerja secara sempurna.

```python
# Verifikasi Numerik Ekuivalensi Anuitas Kasus k > 1 (Due vs Immediate)

i = 0.10  # Bunga tahunan 10%
k = 3     # Pembayaran setiap 3 tahun
n = 6     # Total 6 tahun (2 kali bayar)
R = 1000  # Pembayaran $1000

v = 1 / (1 + i)
a_n = (1 - v**n) / i
a_k = (1 - v**k) / i
s_k = ((1 + i)**k - 1) / i

# Kasus Annuity-Due (Bayar di t=0 dan t=3)
pv_due_rumus = R * (a_n / a_k)
pv_due_manual = R * (1 + v**3)
cicilan_pengganti_due = R / a_k
pv_due_cicilan = cicilan_pengganti_due * a_n

print("=== PEMBAYARAN DI AWAL INTERVAL (ANNUITY-DUE, k > 1) ===")
print(f"Cicilan pengganti per tahun (R / a_k) : ${cicilan_pengganti_due:.2f}")
print(f"PV dengan Rumus (R * a_n / a_k)       : ${pv_due_rumus:.2f}")
print(f"PV dengan Diskonto Manual (1000+1000v^3): ${pv_due_manual:.2f}")
print(f"PV dari Anuitas Cicilan Pengganti      : ${pv_due_cicilan:.2f}")
```
