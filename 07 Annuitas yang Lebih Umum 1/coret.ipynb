### B. Skenario Konkret di Dunia Nyata: "Tabungan Rencana dengan Kenaikan Gaji"

>- Karena ada selisih tepat 1 kuartal antara setoran terakhir dengan titik evaluasi, formula yang digunakan adalah **Anuitas Awal ($\ddot{s}$)**.

### D. Penurunan Matematis Langkah demi Langkah

#### Langkah 1: Suku Bunga Efektif per Kuartal ($j$)
Karena bunga dibungakan bulanan dengan $i_{\text{bulan}} = 1\% = 0.01$, dalam 1 kuartal (3 bulan) uang bertumbuh sebesar:
$$1 + j = (1 + i_{\text{bulan}})^3 = (1 + 0.01)^3 = 1.030301$$
$$j = 0.030301 \quad (3.0301\% \text{ per kuartal})$$

Tingkat diskon efektif per kuartal ($d_j$) adalah:
$$d_j = \frac{j}{1+j} = \frac{0.030301}{1.030301} \approx 0.0294099$$

---

#### Langkah 2: Dua Cara Pandang Menyelesaikan Arus Kas

##### Cara 1: Metode Pelapisan Anuitas (*Layering Method*)
Kita dapat memandang tabungan Budi sebagai dua lapis anuitas:
1. **Lapisan Dasar (Layer 1):** Setoran $\$100$ yang berjalan penuh sepanjang 16 kuartal ($t=0$ s.d. $15$).
2. **Lapisan Tambahan (Layer 2):** Tambahan setoran $\$100$ (karena $\$200 - \$100 = \$100$) yang berjalan selama 8 kuartal terakhir ($t=8$ s.d. $15$).

$$\begin{aligned}
AV_{16} &= 100 \cdot \ddot{s}_{\overline{16}\vert j} + 100 \cdot \ddot{s}_{\overline{8}\vert j} \\[0.5em]
&= 100 \left[ \frac{(1.030301)^{16} - 1}{0.0294099} \right] + 100 \left[ \frac{(1.030301)^8 - 1}{0.0294099} \right] \\[0.5em]
&= 100 \left[ \frac{1.6122261 - 1}{0.0294099} \right] + 100 \left[ \frac{1.2697346 - 1}{0.0294099} \right] \\[0.5em]
&= 100 (20.81704) + 100 (9.17157) \\[0.5em]
&= \$2{,}081.70 + \$917.16 = \mathbf{\$2{,}998.86}
\end{aligned}$$

---

##### Cara 2: Metode Kronologis Berurutan (*Sequential Accumulation*)
Kita juga bisa menghitung saldo rekening sesuai urutan waktu berjalannya hidup Budi:

1. **Hitung Saldo di Akhir Tahun ke-2 ($t = 8$):**
   Akumulasi 8 setoran pertama sebesar $\$100$:
   $$AV_8 = 100 \cdot \ddot{s}_{\overline{8}\vert j} = 100 \times 9.17157 = \$917.16$$

2. **Akumulasikan Saldo Tersebut 2 Tahun ke Depan ($t = 8 \to t = 16$):**
   Saldo $\$917.16$ tersebut dibiarkan terus berbunga selama 8 kuartal berikutnya:
   $$\text{Nilai Pertumbuhan Saldo Lama} = \$917.16 \times (1+j)^8 = \$917.16 \times 1.269735 = \mathbf{\$1{,}164.55}$$

3. **Hitung Akumulasi dari 8 Setoran Baru ($\$200$) di Tahun 3 & 4:**
   $$\text{Nilai Setoran Baru} = 200 \cdot \ddot{s}_{\overline{8}\vert j} = 200 \times 9.17157 = \mathbf{\$1{,}834.31}$$

4. **Total Saldo Akhir di $t = 16$:**
   $$AV_{16} = \$1{,}164.55 + \$1{,}834.31 = \mathbf{\$2{,}998.86}$$

> Kedua metode menghasilkan angka yang **100% identik**, membuktikan konsistensi hukum nilai waktu dari uang (*Time Value of Money*).

```python
# Verifikasi Komputasi dengan Python
# Simulasi Perjalanan Saldo Rekening Budi Tiap Kuartal

i_month = 0.12 / 12  # 1% per bulan
j = (1 + i_month)**3 - 1  # Suku bunga efektif per kuartal
d_j = j / (1 + j)         # Tingkat diskon efektif per kuartal

print(f"Suku bunga efektif per kuartal (j) : {j:.6%}")
print(f"Tingkat diskon per kuartal (d_j)   : {d_j:.6%}")

# Simulasi Mutasi Rekening Buku Tabungan
balance = 0.0
print("\n--- BUKU TABUNGAN REKENING BUDI (PER KUARTAL) ---")
print(f"{'Kuartal':<8}{'Setoran':<12}{'Saldo Awal':<15}{'Bunga Kuartal':<15}{'Saldo Akhir':<15}")

for q in range(16):
    deposit = 100.0 if q < 8 else 200.0
    start_balance = balance + deposit
    interest = start_balance * j
    balance = start_balance + interest
    print(f"{q+1:<8}{deposit:<12.2f}{start_balance:<15.2f}{interest:<15.2f}{balance:<15.2f}")

print("-" * 65)
print(f"TOTAL SALDO AKHIR DI AKHIR TAHUN KE-4: ${balance:.2f}")
```

## 2. Cara Menentukan Suku Bunga Efektif ($j$) dari Bunga Nominal: Mengapa Letak Pangkat Berbeda pada Contoh 1 dan Contoh 2?

### Pertanyaan
> *"Saya kurang paham bagaimana cara menuliskan suku bunga efektif jika diketahui bunga nominalnya. Saya liat pada contoh soal 1 dan contoh soal 2 berbeda mana yang di pangkatkan"*

---

### A. Prinsip Dasar: Apa Target yang Ingin Dicari?

Ketika menyelesaikan soal anuitas umum dengan Pendekatan Numerik:
1. **Target ($j$):** Kita selalu ingin mencari suku bunga efektif per **periode pembayaran transaksi** (misalnya per bulan, per kuartal, per semester, dll).
2. **Modal Dasar ($i_{\text{dasar}}$):** Bunga nominal $i^{(m)}$ selalu dibagi dengan frekuensi konversinya ($m$) untuk mendapatkan bunga efektif per periode pemajemukan bank:
   $$i_{\text{dasar}} = \frac{i^{(m)}}{m}$$

> **Pertanyaan Kuncinya:**
> *"Bagaimana menghubungkan $j$ (periode pembayaran) dengan $i_{\text{dasar}}$ (periode bunga bank)?"*

---

### B. Mengapa Pangkatnya Berbeda? (Perbandingan Panjang Durasi)

Perhatikan perbandingan durasi waktu pada kedua contoh soal:

#### 1. Kasus Contoh 1: Periode Pembayaran LEBIH PANJANG daripada Periode Bunga
- **Periode Pembayaran Transaksi:** Kuartalan (**3 bulan**).
- **Periode Pemajemukan Bank:** Bulanan (**1 bulan**), dengan $i_{\text{bulan}} = \frac{12\%}{12} = 1\% = 0.01$.

**Logika Pertumbuhan Uang:**
- Satu periode pembayaran (3 bulan) menampung **3 kali putaran bunga bulanan**.
- Jadi, pertumbuhan uang selama 1 kuartal ($1+j$) adalah hasil perkalian bunga bulanan sebanyak 3 kali berturut-turut:
  $$\mathbf{1 + j = (1 + i_{\text{bulan}})^3}$$
  $$1 + j = (1 + 0.01)^3 = 1.030301 \implies \mathbf{j = (1.01)^3 - 1 = 0.030301}$$
- 👉 *Pangkat $3$ (bilangan bulat $> 1$) berada di sisi suku bunga bulanan $(1+i_{\text{bulan}})$.*

---

#### 2. Kasus Contoh 2: Periode Pembayaran LEBIH PENDEK daripada Periode Bunga
- **Periode Pembayaran Transaksi:** Kuartalan (**3 bulan**).
- **Periode Pemajemukan Bank:** Semesteran (**6 bulan**), dengan $i_{\text{semester}} = \frac{10\%}{2} = 5\% = 0.05$.

**Logika Pertumbuhan Uang:**
- Satu periode bunga bank (6 bulan) menampung **2 periode pembayaran kuartal**.
- Jadi, jika uang bertumbuh dengan bunga kuartalan $j$ sebanyak 2 kali berturut-turut, hasilnya harus tepat sama dengan pertumbuhan 1 semester:
  $$\mathbf{(1 + j)^2 = 1 + i_{\text{semester}}}$$
  $$(1 + j)^2 = 1 + 0.05 = 1.05$$
- Untuk mencari $(1+j)$, kita tarik akar (pangkatkan kedua ruas dengan $\frac{1}{2}$):
  $$\mathbf{1 + j = (1.05)^{1/2} = \sqrt{1.05}} \implies \mathbf{j = (1.05)^{1/2} - 1 \approx 0.024695}$$
- 👉 *Pangkat $2$ berada di sisi $(1+j)$, yang ketika dipindahkan ke ruas kanan berubah menjadi pangkat pecahan $\frac{1}{2}$.*

### C. Metode Universal Anti-Bingung: Formula Penyetaraan 1 Tahun (*Golden Rule*)

Jika Anda tidak ingin menghafal kasus mana yang lebih panjang atau pendek, gunakan **Satu Prinsip Emas Universal** ini:

> **"Nilai akumulasi dari modal $\$1$ setelah mengendap selama 1 TAHUN PENUH harus sama persis di kedua sudut pandang."**

Definisikan:
- **$p$** = Frekuensi pembayaran transaksi dalam 1 tahun ($p = 4$ untuk kuartal, $p = 12$ untuk bulan, $p = 2$ untuk semester, $p = 1$ untuk tahunan).
- **$m$** = Frekuensi pemajemukan bunga nominal bank dalam 1 tahun ($m$ dari simbol $i^{(m)}$).

Maka dalam rentang waktu **1 tahun**:
$$\underbrace{(1 + j)^p}_{\text{Akumulasi setahun dari } p \text{ periode bayar}} = \underbrace{\left( 1 + \frac{i^{(m)}}{m} \right)^m}_{\text{Akumulasi setahun dari } m \text{ periode konversi bank}}$$

Tarik akar ke-$p$ (pangkatkan kedua ruas dengan $\frac{1}{p}$):

$$\mathbf{1 + j = \left( 1 + \frac{i^{(m)}}{m} \right)^{\frac{m}{p}}}$$

$$\mathbf{j = \left( 1 + \frac{i^{(m)}}{m} \right)^{\frac{m}{p}} - 1}$$

---

### D. Uji Coba Rumus Universal pada Kedua Contoh Soal

Mari kita buktikan betapa mudahnya menggunakan formula universal $\mathbf{\frac{m}{p}}$:

| Komponen | Contoh Soal 1 | Contoh Soal 2 |
| :--- | :--- | :--- |
| **Frekuensi Bayar ($p$)** | Kuartalan $\implies \mathbf{p = 4}$ | Kuartalan $\implies \mathbf{p = 4}$ |
| **Frekuensi Bunga ($m$)** | Bulanan $\implies \mathbf{m = 12}$ | Semesteran $\implies \mathbf{m = 2}$ |
| **Bunga Dasar $\left(\frac{i^{(m)}}{m}\right)$** | $\frac{12\%}{12} = 1\% = 0.01$ | $\frac{10\%}{2} = 5\% = 0.05$ |
| **Pangkat Rasio $\left(\frac{m}{p}\right)$** | $\dfrac{m}{p} = \dfrac{12}{4} = \mathbf{3}$ | $\dfrac{m}{p} = \dfrac{2}{4} = \mathbf{\dfrac{1}{2} = 0.5}$ |
| **Persamaan $j$** | $j = (1 + 0.01)^{\mathbf{3}} - 1$ | $j = (1 + 0.05)^{\mathbf{1/2}} - 1$ |
| **Hasil Akhir $j$** | $\mathbf{3.030101\%}$ per kuartal | $\mathbf{2.469508\%}$ per kuartal |

> **Ringkasan Praktis:**
>- Jika $\frac{m}{p} > 1$ (seperti Contoh 1: $\frac{12}{4} = 3$), pangkatnya berupa bilangan bulat $> 1$.
>- Jika $\frac{m}{p} < 1$ (seperti Contoh 2: $\frac{2}{4} = 0.5$), pangkatnya berupa pecahan (akar).

```python
# Kalkulator Universal Suku Bunga Efektif per Periode Pembayaran (j)

def hitung_suku_bunga_efektif(i_nominal, m, p):
    """
    Menghitung suku bunga efektif per periode pembayaran (j)
    Parameter:
      i_nominal : Suku bunga nominal tahunan (misal 0.12 untuk 12%)
      m         : Frekuensi pemajemukan bunga per tahun (misal 12 untuk bulanan, 2 untuk semesteran)
      p         : Frekuensi pembayaran transaksi per tahun (misal 4 untuk kuartalan, 12 untuk bulanan)
    """
    i_dasar = i_nominal / m
    rasio_pangkat = m / p
    j = (1 + i_dasar)**rasio_pangkat - 1
    return i_dasar, rasio_pangkat, j

# Uji Kasus Contoh 1: i^(12) = 12%, bayar kuartalan (p=4)
i_dasar1, rasio1, j1 = hitung_suku_bunga_efektif(0.12, m=12, p=4)
print("=== CONTOH SOAL 1 ===")
print(f"Bunga per periode bank (bulanan) : {i_dasar1:.4%}")
print(f"Pangkat rasio (m/p = 12/4)       : {rasio1:.1f}")
print(f"Suku bunga efektif per kuartal (j): {j1:.6%}")

# Uji Kasus Contoh 2: i^(2) = 10%, bayar kuartalan (p=4)
i_dasar2, rasio2, j2 = hitung_suku_bunga_efektif(0.10, m=2, p=4)
print("\n=== CONTOH SOAL 2 ===")
print(f"Bunga per periode bank (semester): {i_dasar2:.4%}")
print(f"Pangkat rasio (m/p = 2/4)        : {rasio2:.2f}")
print(f"Suku bunga efektif per kuartal (j): {j2:.6%}")
```

## 3. Interpretasi Rumus Pokok: Pembayaran 1 Setiap Awal $k$ Periode Bunga ($PV = \frac{a_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}$)

### Pertanyaan
> *"Saya tidak paham interpretasi pada 'Rumus Pokok: Pembayaran 1 Setiap Awal k Periode Bunga' (Catatan 07 Subbab 3.3)"*

---

### A. Bunyi Rumus & Pernyataan dalam Catatan 07

Pada Subbab 3.3, untuk transaksi di mana uang sebesar **$\$1$** dibayarkan pada **setiap awal $k$ periode bunga** ($t = 0, k, 2k, \dots, n-k$):
$$\mathbf{PV = \frac{a_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}} \qquad \text{dan} \qquad \mathbf{FV = \frac{s_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}}$$

> **Interpretasi Aktuarial:**
> *"Pembayaran $1$ pada awal setiap interval sepanjang $k$ periode bunga dapat digantikan oleh pembayaran berkala sebesar $\frac{1}{a_{\overline{k}\vert i}}$ pada setiap akhir periode bunga tunggal."*

---

### B. Mengapa Bisa Digantikan? (Intuisi Finansial Konkret)

Mari kita bedah secara logis bagaimana uang $\$1$ di awal interval bisa digantikan dengan cicilan $\frac{1}{a_{\overline{k}\vert i}}$:

#### 1. Tinjau 1 Blok Interval Saja ($t = 0$ sampai $t = k$)
- Bayangkan pada saat **$t = 0$ (hari ini)**, Anda menerima uang tunai sebesar **$\$1$**.
- Menerima uang tunai $\$1$ di awal hari ini secara finansial ekuivalen dengan **mengambil pinjaman sebesar $\$1$** di $t=0$.
- Jika Anda memiliki utang $\$1$ di $t=0$ dan ingin melunasinya dengan mencicil tetap sebesar $C$ di **setiap akhir periode** selama $k$ periode berturut-turut ($t = 1, 2, \dots, k$), berapa besarnya cicilan $C$?
  $$1 = C \cdot a_{\overline{k}\vert i} \implies \mathbf{C = \frac{1}{a_{\overline{k}\vert i}}}$$

> **Kesimpulan Blok 1:**
> Menerima uang tunai **$\$1$ sekaligus di awal ($t=0$)** memiliki nilai ekonomi yang **100% SAMA** dengan menerima **aliran cicilan sebesar $C = \frac{1}{a_{\overline{k}\vert i}}$ di setiap $t = 1, 2, \dots, k$**.

---

#### 2. Terapkan Prinsip Penggantian Ini ke Seluruh Blok Pembayaran

Jika kita ganti setiap setoran $\$1$ gelondongan dengan serangkaian cicilan ekuivalennya:
1. **Setoran $\$1$ di $t=0$** $\longrightarrow$ Diganti cicilan $C = \frac{1}{a_{\overline{k}\vert i}}$ pada $t = 1, 2, \dots, k$.
2. **Setoran $\$1$ di $t=k$** $\longrightarrow$ Diganti cicilan $C = \frac{1}{a_{\overline{k}\vert i}}$ pada $t = k+1, k+2, \dots, 2k$.
3. **Setoran $\$1$ di $t=2k$** $\longrightarrow$ Diganti cicilan $C = \frac{1}{a_{\overline{k}\vert i}}$ pada $t = 2k+1, \dots, 3k$.
4. $\dots$
5. **Setoran $\$1$ di $t=n-k$** $\longrightarrow$ Diganti cicilan $C = \frac{1}{a_{\overline{k}\vert i}}$ pada $t = n-k+1, \dots, n$.

---

#### 3. Garis Waktu Baru yang Tercipta

Perhatikan apa yang terjadi pada seluruh garis waktu dari $t = 1$ hingga $t = n$:
- Pembayaran gelondongan yang jarang-jarang (setiap $k$ periode) kini telah **lebur/terdistribusi merata** menjadi **pembayaran konstan sebesar $C = \frac{1}{a_{\overline{k}\vert i}}$ pada SETIAP akhir periode bunga tanpa jeda** dari $t = 1$ sampai $t = n$.
- Rangkaian pembayaran baru ini tidak lain adalah **Anuitas Akhir Biasa (*Standard Annuity-Immediate*)** berjangka $n$ periode dengan besar pembayaran $C$!

Maka Nilai Sekarang ($PV$) dari seluruh transaksi adalah:
$$PV = C \cdot a_{\overline{n}\vert i} = \left( \frac{1}{a_{\overline{k}\vert i}} \right) \cdot a_{\overline{n}\vert i} = \mathbf{\frac{a_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}}$$

Dan Nilai Terakumulasi ($FV$) pada $t = n$ adalah:
$$FV = C \cdot s_{\overline{n}\vert i} = \left( \frac{1}{a_{\overline{k}\vert i}} \right) \cdot s_{\overline{n}\vert i} = \mathbf{\frac{s_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}}$$

> Inilah rahasia mengapa simbol $a_{\overline{k}\vert i}$ muncul sebagai penyebut pembagi!

### C. Perbandingan Kontras: Anuitas Akhir vs Anuitas Awal ($k > 1$)

Untuk melihat keindahan simetrisnya, mari bandingkan kedua kasus pembagian ini:

| Karakteristik | Kasus Anuitas Akhir ($k > 1$) | Kasus Anuitas Awal ($k > 1$) |
| :--- | :--- | :--- |
| **Titik Pembayaran** | Di **AKHIR** interval ($t = k, 2k, \dots, n$) | Di **AWAL** interval ($t = 0, k, \dots, n-k$) |
| **Analogi Keuangan** | *Sinking Fund* (menabung $C$ tiap periode untuk mengumpulkan $\$1$ di **akhir** periode ke-$k$) | *Loan Amortization* (mencicil $C$ tiap periode untuk melunasi utang $\$1$ yang diterima di **awal**) |
| **Nilai Cicilan Pengganti ($C$)** | $1 = C \cdot s_{\overline{k}\vert i} \implies \mathbf{C = \dfrac{1}{s_{\overline{k}\vert i}}}$ | $1 = C \cdot a_{\overline{k}\vert i} \implies \mathbf{C = \dfrac{1}{a_{\overline{k}\vert i}}}$ |
| **Pembagi Formula** | Dibagi **$s_{\overline{k}\vert i}$** | Dibagi **$a_{\overline{k}\vert i}$** |
| **Rumus Nilai Sekarang ($PV$)** | $\mathbf{PV = \dfrac{a_{\overline{n}\vert i}}{s_{\overline{k}\vert i}}}$ | $\mathbf{PV = \dfrac{a_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}}$ |
| **Rumus Nilai Akumulasi ($FV$)** | $\mathbf{FV = \dfrac{s_{\overline{n}\vert i}}{s_{\overline{k}\vert i}}}$ | $\mathbf{FV = \dfrac{s_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}}$ |

---

### D. Pembuktian Aljabar Murni Deret Geometri

Jika Anda ingin melihat penurunan matematis murninya tanpa analogi cicilan:
- Arus kas terjadi di $t = 0, k, 2k, \dots, n-k$ sebanyak $m = n/k$ kali pembayaran.
- Diskontokan masing-masing ke $t = 0$:
  $$PV = 1 + v^k + v^{2k} + \dots + v^{n-k}$$
- Jumlah deret geometri dengan rasio $r = v^k$ dan $m$ suku:
  $$PV = \frac{1 - (v^k)^m}{1 - v^k} = \frac{1 - v^n}{1 - v^k}$$
- Bagi pembilang dan penyebut dengan $i$:
  $$PV = \frac{\dfrac{1 - v^n}{i}}{\dfrac{1 - v^k}{i}} = \mathbf{\frac{a_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}}$$

---

### E. Contoh Angka Nyata

Misalkan suku bunga $i = 10\%$ per tahun. Anda dijanjikan menerima **$\$1{,}000$ di awal setiap 3 tahun ($k = 3$)** selama total jangka waktu **6 tahun ($n = 6$)**.
- Artinya, Anda menerima $\$1{,}000$ di $t = 0$ dan $\$1{,}000$ di $t = 3$.

**Cara 1 (Konsep Penggantian Cicilan):**
Uang $\$1{,}000$ di awal setiap 3 tahun digantikan oleh cicilan tahunan:
$$C = \frac{1000}{a_{\overline{3}\vert 0.10}} = \frac{1000}{2.48685} = \$402.11 \text{ per tahun selama 6 tahun}$$
Maka $PV = 402.11 \times a_{\overline{6}\vert 0.10} = 402.11 \times 4.35526 = \mathbf{\$1{,}751.31}$

**Cara 2 (Rumus Langsung $\frac{a_{\overline{n}\vert i}}{a_{\overline{k}\vert i}}$):**
$$PV = 1000 \cdot \frac{a_{\overline{6}\vert 0.10}}{a_{\overline{3}\vert 0.10}} = 1000 \times \frac{4.35526}{2.48685} = \mathbf{\$1{,}751.31}$$

**Cara 3 (Diskonto Manual Kas Tunggal):**
$$PV = 1000 + 1000(1.10)^{-3} = 1000 + 751.31 = \mathbf{\$1{,}751.31}$$

> Ketiga cara menghasilkan angka yang **100% sama**, membuktikan bahwa interpretasi penggantian cicilan $1/a_{\overline{k}\vert i}$ benar-benar bekerja secara sempurna.

```python
# Verifikasi Numerik Ekuivalensi Anuitas Kasus k > 1 (Due vs Immediate)

i = 0.10  # Bunga tahunan 10%
k = 3     # Pembayaran setiap 3 tahun
n = 6     # Total 6 tahun (2 kali bayar)
R = 1000  # Pembayaran $1000

v = 1 / (1 + i)
a_n = (1 - v**n) / i
a_k = (1 - v**k) / i
s_k = ((1 + i)**k - 1) / i

# Kasus Annuity-Due (Bayar di t=0 dan t=3)
pv_due_rumus = R * (a_n / a_k)
pv_due_manual = R * (1 + v**3)
cicilan_pengganti_due = R / a_k
pv_due_cicilan = cicilan_pengganti_due * a_n

print("=== PEMBAYARAN DI AWAL INTERVAL (ANNUITY-DUE, k > 1) ===")
print(f"Cicilan pengganti per tahun (R / a_k) : ${cicilan_pengganti_due:.2f}")
print(f"PV dengan Rumus (R * a_n / a_k)       : ${pv_due_rumus:.2f}")
print(f"PV dengan Diskonto Manual (1000+1000v^3): ${pv_due_manual:.2f}")
print(f"PV dari Anuitas Cicilan Pengganti      : ${pv_due_cicilan:.2f}")
```

## 4. Penjelasan & Penurunan Matematis Kasus 2: Pembayaran Lebih Sering daripada Pemajemukan Bunga ($m > 1$)

### Pertanyaan
> *"Saya tidak paham dengan penurunan matematis untuk kasus 2"*

---

### A. Konsep Dasar & Notasi Kasus 2

Pada **Kasus 2**, pembayaran terjadi **lebih sering** daripada pemajemukan bunga:
- **Unit Waktu Acuan:** 1 periode bunga (misalnya **1 tahun**), dengan suku bunga efektif per tahun sebesar **$i$** dan faktor diskonto **$v = (1+i)^{-1}$**.
- **Frekuensi Pembayaran:** Sebanyak **$m$ kali dalam setahun** (misal $m = 12$ kali / bulanan).
- **Besar Pembayaran:** Total uang yang dibayarkan dalam 1 tahun didefinisikan sebesar **$1$ satuan**, sehingga setiap kali bayar besarnya adalah **$\frac{1}{m}$** (misal $\$1/12$).
- **Jangka Waktu:** Berlangsung selama **$n$ tahun**, sehingga total banyaknya pembayaran adalah **$mn$ kali**.

---

### B. Penurunan Matematis Nilai Sekarang Anuitas Akhir ($a_{\overline{n}\vert}^{(m)}$)

#### 1. Garis Waktu Arus Kas
Pembayaran sebesar $\frac{1}{m}$ terjadi di setiap akhir sub-periode $\frac{1}{m}$, yaitu pada titik waktu:
$$t = \frac{1}{m}, \frac{2}{m}, \frac{3}{m}, \dots, \frac{mn}{m} = n$$

#### 2. Menuliskan Deret Nilai Sekarang ($PV$ di $t = 0$)
Setiap pembayaran sebesar $\frac{1}{m}$ pada waktu $t$ ditarik mundur ke $t = 0$ dengan faktor diskonto $v^t$:
$$a_{\overline{n}\vert}^{(m)} = \frac{1}{m} v^{1/m} + \frac{1}{m} v^{2/m} + \frac{1}{m} v^{3/m} + \dots + \frac{1}{m} v^n$$

Faktorkan keluar konstanta $\frac{1}{m}$:
$$a_{\overline{n}\vert}^{(m)} = \frac{1}{m} \left( v^{1/m} + v^{2/m} + v^{3/m} + \dots + v^n \right)$$

#### 3. Menjumlahkan Deret Geometri
Deret di dalam tanda kurung adalah deret geometri terhingga:
- Suku pertama: $a_1 = v^{1/m}$
- Rasio pengali: $r = v^{1/m}$
- Banyaknya suku: $N = mn$

Gunakan rumus jumlah deret geometri $S_N = a_1 \cdot \dfrac{1 - r^N}{1 - r}$:
$$S_{mn} = v^{1/m} \cdot \left[ \frac{1 - (v^{1/m})^{mn}}{1 - v^{1/m}} \right] = v^{1/m} \cdot \left[ \frac{1 - v^n}{1 - v^{1/m}} \right]$$

#### 4. Manipulasi Aljabar Pecahan
Bagi pembilang dan penyebut pecahan tersebut dengan $v^{1/m}$ (atau kalikan dengan $v^{-1/m} = (1+i)^{1/m}$):
$$v^{1/m} \cdot \left[ \frac{1 - v^n}{1 - v^{1/m}} \right] = \frac{1 - v^n}{v^{-1/m} \left( 1 - v^{1/m} \right)} = \frac{1 - v^n}{v^{-1/m} - 1} = \frac{1 - v^n}{(1+i)^{1/m} - 1}$$

Substitusikan kembali ke persamaan $a_{\overline{n}\vert}^{(m)}$:
$$a_{\overline{n}\vert}^{(m)} = \frac{1}{m} \cdot \left[ \frac{1 - v^n}{(1+i)^{1/m} - 1} \right] = \frac{1 - v^n}{m \left[ (1+i)^{1/m} - 1 \right]}$$

#### 5. Momen Kunci: Hubungan dengan Suku Bunga Nominal ($i^{(m)}$)
Ingat kembali definisi suku bunga nominal tahunan yang dikonversi $m$ kali setahun (Bab 02):
$$i^{(m)} = m \left[ (1+i)^{1/m} - 1 \right]$$

Penyebut pecahan kita persis adalah $i^{(m)}$! Maka terbukti secara elegan:
$$\mathbf{a_{\overline{n}\vert}^{(m)} = \frac{1 - v^n}{i^{(m)}}}}$$

#### 6. Menghubungkan dengan Anuitas Tahunan Biasa ($a_{\overline{n}\vert i}$)
Karena rumus anuitas biasa adalah $a_{\overline{n}\vert i} = \dfrac{1 - v^n}{i} \implies 1 - v^n = i \cdot a_{\overline{n}\vert i}$, kita peroleh:
$$\mathbf{a_{\overline{n}\vert}^{(m)} = \frac{i \cdot a_{\overline{n}\vert i}}{i^{(m)}} = \left( \frac{i}{i^{(m)}} \right) \cdot a_{\overline{n}\vert i}}$$

---

### C. Penurunan Matematis Nilai Sekarang Anuitas Awal ($\ddot{a}_{\overline{n}\vert}^{(m)}$)

#### 1. Garis Waktu Arus Kas
Pembayaran sebesar $\frac{1}{m}$ dilakukan di **awal** setiap sub-periode $\frac{1}{m}$, yaitu pada:
$$t = 0, \frac{1}{m}, \frac{2}{m}, \dots, \frac{mn-1}{m} = n - \frac{1}{m}$$

#### 2. Menuliskan Deret Nilai Sekarang
$$\ddot{a}_{\overline{n}\vert}^{(m)} = \frac{1}{m} \left( 1 + v^{1/m} + v^{2/m} + \dots + v^{n - 1/m} \right)$$

#### 3. Jumlah Deret Geometri
- Suku pertama: $a_1 = 1$
- Rasio: $r = v^{1/m}$
- Banyak suku: $mn$

$$\ddot{a}_{\overline{n}\vert}^{(m)} = \frac{1}{m} \cdot \left[ \frac{1 - (v^{1/m})^{mn}}{1 - v^{1/m}} \right] = \frac{1 - v^n}{m \left( 1 - v^{1/m} \right)}$$

#### 4. Momen Kunci: Hubungan dengan Tingkat Diskon Nominal ($d^{(m)}$)
Ingat kembali definisi tingkat diskon nominal tahunan (Bab 02):
$$d^{(m)} = m \left( 1 - v^{1/m} \right)$$

Penyebutnya persis adalah $d^{(m)}$! Maka diperoleh:
$$\mathbf{\ddot{a}_{\overline{n}\vert}^{(m)} = \frac{1 - v^n}{d^{(m)}} = \left( \frac{i}{d^{(m)}} \right) \cdot a_{\overline{n}\vert i} = \left( \frac{d}{d^{(m)}} \right) \cdot \ddot{a}_{\overline{n}\vert i}}$$

---

### D. Penurunan Nilai Akumulasi ($s_{\overline{n}\vert}^{(m)}$ dan $\ddot{s}_{\overline{n}\vert}^{(m)}$)

Untuk mencari nilai di titik akhir $t = n$, kita cukup mengalikan nilai sekarang dengan faktor akumulasi $(1+i)^n$:

1. **Anuitas Akhir:**
   $$s_{\overline{n}\vert}^{(m)} = a_{\overline{n}\vert}^{(m)} \cdot (1+i)^n = \left( \frac{1 - v^n}{i^{(m)}} \right) (1+i)^n = \mathbf{\frac{(1+i)^n - 1}{i^{(m)}} = \left( \frac{i}{i^{(m)}} \right) s_{\overline{n}\vert i}}$$

2. **Anuitas Awal:**
   $$\ddot{s}_{\overline{n}\vert}^{(m)} = \ddot{a}_{\overline{n}\vert}^{(m)} \cdot (1+i)^n = \left( \frac{1 - v^n}{d^{(m)}} \right) (1+i)^n = \mathbf{\frac{(1+i)^n - 1}{d^{(m)}} = \left( \frac{i}{d^{(m)}} \right) s_{\overline{n}\vert i}}$$

---

### E. Intuisi Finansial: Mengapa Muncul Pengali $\frac{i}{i^{(m)}}$?

Bayangkan Anda memiliki dua opsi untuk menerima total uang $\$1{,}200$ dalam 1 tahun ($i = 10\%$):
- **Opsi A (Tahunan, $a_{\overline{n}\vert i}$):** Menerima $\$1{,}200$ sekaligus di **akhir tahun** ($t=1$).
- **Opsi B (Bulanan, $a_{\overline{n}\vert}^{(12)}$):** Menerima **$\$100$ di setiap akhir bulan** (Jan, Feb, Mar, $\dots$, Des).

> **Pertanyaan Logis:** Mana yang lebih menguntungkan?
> **Jawaban:** Tentu saja **Opsi B**! Karena Anda sudah mulai menerima uang sejak bulan Januari, uang tersebut bisa langsung Anda investasikan kembali untuk memupuk bunga sepanjang tahun berjalan.

Karena uang diterima **lebih awal**, nilai sekarang dari arus kas bulanan **pasti lebih besar** daripada arus kas tahunan:
$$i^{(m)} < i \implies \frac{i}{i^{(m)}} > 1 \implies \mathbf{a_{\overline{n}\vert}^{(m)} > a_{\overline{n}\vert i}}$$

Faktor pengali $\mathbf{\dfrac{i}{i^{(m)}}}$ adalah "faktor penyesuaian bunga tambahan" yang mencerminkan keuntungan dari penerimaan uang lebih awal di sepanjang tahun!

```python
# Verifikasi Penurunan Rumus Kasus 2 dengan Python

i = 0.08      # Suku bunga efektif tahunan 8%
n = 5         # Jangka waktu 5 tahun
m = 12        # Pembayaran bulanan (m=12)

v = 1 / (1 + i)
i_m = m * ((1 + i)**(1/m) - 1)  # Suku bunga nominal i^(12)
d_m = m * (1 - v**(1/m))         # Tingkat diskon nominal d^(12)

# 1. Hitung langsung dari deret geometri eksplisit
deret_pv = sum((1/m) * (v**(t/m)) for t in range(1, m*n + 1))

# 2. Hitung menggunakan rumus analitis (1 - v^n) / i^(m)
rumus_pv = (1 - v**n) / i_m

# 3. Hitung menggunakan faktor pengali (i / i^(m)) * a_n
a_n = (1 - v**n) / i
pengali_pv = (i / i_m) * a_n

print("=== HASIL PENURUNAN RUMUS a_n^(m) ===")
print(f"1. Jumlahan Deret Eksplisit : {deret_pv:.8f}")
print(f"2. Rumus (1 - v^n) / i^(m)  : {rumus_pv:.8f}")
print(f"3. Rumus (i / i^(m)) * a_n  : {pengali_pv:.8f}")
print(f"Faktor Pengali (i / i^(m))  : {i / i_m:.6f} (> 1, karena uang diterima lebih awal)")
```

## 5. Mengapa pada $a_{\overline{n}\vert}^{(m)}$, Faktor Diskon $v$ Dipangkatkan $\frac{1}{m}$ dan Kelipatannya?

### A. Penjelasan 1: Satuan Waktu Acuan Adalah TAHUNAN, Bukan Sub-Periode

1. **Definisi Faktor Diskonto $v$:**
   $$v = (1+i)^{-1}$$
   Di mana $i$ adalah suku bunga efektif per **1 TAHUN PENUH**.
   Artinya, mengalikan sejumlah uang dengan $v^t$ berarti **menarik mundur nilai uang tersebut sejauh rentang waktu $t$ TAHUN**:
   $$\text{Nilai Sekarang dari } \$1 \text{ pada waktu } t \text{ tahun} = 1 \cdot v^t = (1+i)^{-t}$$

2. **Kapan Pembayaran Terjadi? (Posisi Waktu dalam Satuan Tahun):**
   Karena dalam 1 tahun ada $m$ kali pembayaran (misal bulanan, $m = 12$):
   - Pembayaran ke-1 (akhir bulan ke-1) $\implies t = \mathbf{\frac{1}{12}\text{ tahun}} = \mathbf{\frac{1}{m}}$
   - Pembayaran ke-2 (akhir bulan ke-2) $\implies t = \mathbf{\frac{2}{12}\text{ tahun}} = \mathbf{\frac{2}{m}}$
   - Pembayaran ke-3 (akhir bulan ke-3) $\implies t = \mathbf{\frac{3}{12}\text{ tahun}} = \mathbf{\frac{3}{m}}$
   - $\dots$
   - Pembayaran ke-$k$ (akhir bulan ke-$k$) $\implies t = \mathbf{\frac{k}{m}\text{ tahun}}$
   - $\dots$
   - Pembayaran ke-$m$ (akhir bulan ke-12 / akhir tahun ke-1) $\implies t = \frac{m}{m} = \mathbf{1\text{ tahun}}$
   - $\dots$
   - Pembayaran ke-$mn$ (akhir tahun ke-$n$) $\implies t = \frac{mn}{m} = \mathbf{n\text{ tahun}}$

3. **Mendiskontokan Masing-Masing Pembayaran:**
   Karena uang sebesar $\frac{1}{m}$ jatuh tempo pada titik-titik waktu $t = \frac{1}{m}, \frac{2}{m}, \dots, \frac{k}{m}, \dots, n$, maka ketika ditarik mundur ke $t = 0$, pangkat dari $v$ secara alami mengikuti nilai waktu $t$ tersebut:
   $$PV = \frac{1}{m} v^{1/m} + \frac{1}{m} v^{2/m} + \frac{1}{m} v^{3/m} + \dots + \frac{1}{m} v^{k/m} + \dots + \frac{1}{m} v^n$$

### B. Penjelasan 2: Sudut Pandang Faktor Diskonto Sub-Periode ($v_{\text{sub}}$)

Jika kita pandang dari sudut diskonto per 1 sub-periode (misal diskonto bulanan):
- Misalkan $j$ adalah suku bunga efektif per $\frac{1}{m}$ periode (per bulan), maka:
  $$(1+j)^m = 1+i \implies 1+j = (1+i)^{1/m}$$
- Faktor diskonto untuk 1 sub-periode (mundur 1 bulan) adalah:
  $$v_{\text{sub}} = \frac{1}{1+j} = \frac{1}{(1+i)^{1/m}} = (1+i)^{-1/m} = \left[ (1+i)^{-1} \right]^{1/m} = \mathbf{v^{1/m}}$$

Maka perpindahan mundur di garis waktu adalah:
- Mundur 1 sub-periode (1 bulan) $\implies \times v_{\text{sub}} = \mathbf{v^{1/m}}$
- Mundur 2 sub-periode (2 bulan) $\implies \times (v_{\text{sub}})^2 = (v^{1/m})^2 = \mathbf{v^{2/m}}$
- Mundur 3 sub-periode (3 bulan) $\implies \times (v_{\text{sub}})^3 = (v^{1/m})^3 = \mathbf{v^{3/m}}$
- Mundur $k$ sub-periode ($k$ bulan) $\implies \times (v_{\text{sub}})^k = \mathbf{v^{k/m}}$

### D. Perhitungan Nilai Sekarang (*Present Value* pada 1 September 2000)

#### 1. Nilai Sekarang *Fellowship* (Anuitas Awal / *Annuity-Due*)
$$\begin{aligned}
PV_{\text{Fellowship}} &= 5800 \cdot \ddot{a}_{\overline{4}\vert 0.04} \\[0.5em]
&= 5800 \cdot (1.04) \cdot \left[ \frac{1 - (1.04)^{-4}}{0.04} \right] \\[0.5em]
&= 5800 \times 3.775091 = \mathbf{\$21{,}895.53}
\end{aligned}$$

#### 2. Nilai Sekarang *Scholarship* (Anuitas Bulanan / *Annuity-Immediate $m=12$*)
Suku bunga nominal tahunan $i^{(12)}$:
$$i^{(12)} = 12 \left[ (1.04)^{1/12} - 1 \right] \approx 0.03928488$$

$$\begin{aligned}
PV_{\text{Scholarship}} &= 5904 \cdot a_{\overline{4}\vert 0.04}^{(12)} \\[0.5em]
&= 5904 \cdot \frac{1 - (1.04)^{-4}}{i^{(12)}} \\[0.5em]
&= 5904 \times 3.695972 = \mathbf{\$21{,}820.83}
\end{aligned}$$

---

### E. Kesimpulan Keputusan Finansial

$$\mathbf{PV_{\text{Fellowship}} (\$21{,}895.53) > PV_{\text{Scholarship}} (\$21{,}820.83)}$$

- Meskipun total nominal *Fellowship* tampak lebih kecil $\$416$, **nilai ekonomis riil *Fellowship* justru LEBIH BESAR $\$74.70$** dibandingkan *Scholarship*.
- **Mengapa?** Karena keuntungan bunga dari memegang uang $\$5.800$ di depan jauh lebih bernilai daripada menunggu cicilan bulanan yang datang terlambat.

> **Dua Sudut Pandang Utama:**
>- **Bagi Mahpee:** Mengambil *Fellowship* memberikan daya beli dan potensi akumulasi tabungan yang lebih besar di akhir tahun ke-4.
>- **Bagi Lembaga Donor/Yayasan:** Untuk memberikan *Fellowship*, yayasan harus menyisihkan dana abadi lebih besar ($\$21{,}895.53$) hari ini dibanding jika memberikan *Scholarship* ($\$21{,}820.83$).

```python
# Verifikasi Numerik Komparasi Beasiswa Mahpee (Fellowship vs Scholarship)

i = 0.04  # Suku bunga efektif tahunan 4%
n = 4     # Jangka waktu 4 tahun

# 1. Valuasi Fellowship (Annuity-Due tahunan $5800)
v = 1 / (1 + i)
a_due_4 = (1 + i) * ((1 - v**n) / i)
pv_fellowship = 5800 * a_due_4

# 2. Valuasi Scholarship (Annuity-Immediate bulanan, total $5904/tahun)
i_12 = 12 * ((1 + i)**(1/12) - 1)
a_m_4 = (1 - v**n) / i_12
pv_scholarship = 5904 * a_m_4

print("=== HASIL VALUASI NILAI SEKARANG (1 SEPT 2000) ===")
print(f"Total Nominal Mentah Fellowship  : $5,800 x 4  = ${5800*4:,.2f}")
print(f"Total Nominal Mentah Scholarship : $5,904 x 4  = ${5904*4:,.2f}")
print(f"Selisih Nominal (Scholarship menang) : +${5904*4 - 5800*4:,.2f}")
print("-" * 55)
print(f"Nilai Sekarang (PV) Fellowship   : ${pv_fellowship:,.2f}")
print(f"Nilai Sekarang (PV) Scholarship  : ${pv_scholarship:,.2f}")
print(f"Selisih PV Riil (Fellowship menang) : +${pv_fellowship - pv_scholarship:,.2f}")
```

## 7. Latihan Soal 2: Penjelasan Notasi $a_{\overline{10}\vert}^{(12)}$ vs $a_{\overline{120}\vert j}$ dan Cara Mendapatkan Nilai $j$

### Pertanyaan
> *"Saya tidak paham Latihan Soal 2: Penentuan Suku Bunga Nominal dari Nilai Sekarang. Kenapa mencari $a_{\overline{10}\vert}^{(12)}$ dan juga $a_{\overline{120}\vert j}$? Dari mana nilai $j$ didapatkan?"*

---

### A. Bunyi Latihan Soal 2 (Catatan 07 Subbab 5.2)
> Diketahui bahwa nilai sekarang dari anuitas akhir 10 tahun yang membayar **$\$100$ pada setiap akhir bulan** adalah **$\$8.000$**.
> Hitunglah suku bunga nominal tahunan **$i^{(12)}$** yang mendasari transaksi ini!

---

### B. Mengapa Muncul Dua Notasi: $a_{\overline{10}\vert}^{(12)}$ dan $a_{\overline{120}\vert j}$?

Kedua notasi tersebut **bukanlah dua hal yang berbeda**, melainkan **dua cara pandang setara untuk menuliskan persamaan arus kas yang sama persis**:

#### 1. Sudut Pandang 1: Satuan Waktu TAHUNAN (Notasi Aktuaria $a_{\overline{n}\vert}^{(m)}$)
- Satuan waktu acuan: **Tahun** ($n = 10\text{ tahun}$).
- Frekuensi pembayaran: $m = 12\text{ kali setahun}$ (bulanan).
- Total pembayaran per tahun: $12 \times \$100 = \mathbf{\$1.200\text{/tahun}}$.
- Berdasarkan definisi aktuarial, simbol $a_{\overline{n}\vert}^{(m)}$ adalah nilai sekarang dari anuitas dengan **total pembayaran $1$ per tahun** yang dicairkan $m$ kali setahun (sebesar $1/m$ tiap sub-periode).
- Karena total bayar per tahun adalah $\$1.200$, maka persamaan nilai sekarangnya adalah:
  $$PV = 1200 \cdot a_{\overline{10}\vert}^{(12)} = 8000 \implies \mathbf{a_{\overline{10}\vert}^{(12)} = \frac{8000}{1200} = \frac{20}{3} \approx 6.6667}$$

#### 2. Sudut Pandang 2: Satuan Waktu BULANAN (Anuitas Biasa Dasar $a_{\overline{N}\vert j}$)
- Satuan waktu acuan langsung diubah ke: **Bulan**.
- Total banyaknya periode pembayaran: $N = 10\text{ tahun} \times 12\text{ bulan/tahun} = \mathbf{120\text{ bulan}}$.
- Pembayaran per periode: $\mathbf{\$100\text{ per bulan}}$.
- Suku bunga efektif per bulan: $j = \frac{i^{(12)}}{12}$.
- Ini adalah anuitas akhir biasa (*standard annuity-immediate*) dengan 120 kali pembayaran:
  $$PV = 100 \cdot a_{\overline{120}\vert j} = 8000 \implies \mathbf{a_{\overline{120}\vert j} = \frac{8000}{100} = 80.0}$$

#### 3. Jembatan Hubungan Keduanya
Perhatikan bahwa:
$$a_{\overline{10}\vert}^{(12)} = \frac{1}{12} a_{\overline{120}\vert j} \iff 6.6667 = \frac{80.0}{12}$$

> **Mengapa Sudut Pandang 2 ($a_{\overline{120}\vert j} = 80$) yang Dipakai untuk Menghitung?**
> Karena rumus $a_{\overline{120}\vert j} = \frac{1 - (1+j)^{-120}}{j} = 80$ adalah bentuk anuitas dasar yang paling mudah dihitung dengan kalkulator finansial maupun metode interpolasi.

---

### C. Dari Mana Nilai $j$ Didapatkan?

Persamaan yang harus dipecahkan adalah:
$$\frac{1 - (1+j)^{-120}}{j} = 80$$

#### 1. Mengapa Tidak Bisa Diisolasi Langsung Aljabar ($j = \dots$)?
Variabel $j$ berada di penyebut pecahan sekaligus di dalam pangkat $(1+j)^{-120}$. Ini adalah **persamaan polinomial berderajat 121 (persamaan transendental)** yang secara analitis aljabar **tidak memiliki solusi bentuk tertutup (*closed-form solution*)**.

Oleh karena itu, nilai $j$ dicari menggunakan **Metode Numerik** / **Aproksimasi**.

---

#### 2. Tiga Cara Menentukan Nilai $j$

##### Cara 1: Menggunakan Kalkulator Finansial (Standar Ujian PAI / SOA FM - BA II Plus)
Masukkan variabel *Time Value of Money (TVM)* ke kalkulator:
- `N = 120` (Total bulan)
- `PV = -8000` (Uang keluar/masuk di awal)
- `PMT = 100` (Pembayaran bulanan)
- `FV = 0` (Saldo lunas di akhir)
- Tekan `[CPT]` lalu `[I/Y]` $\implies \mathbf{j = 0.7241\% = 0.007241}$ per bulan.

##### Cara 2: Metode Coba-Coba (*Trial and Error*) & Interpolasi Linier (Jika Manual)
Kita cari dua nilai $j$ tebakan yang mengapit target $PV = 80.0$:
- **Tebakan 1:** Misal $j_1 = 0.70\% = 0.0070$:
  $$a_{\overline{120}\vert 0.0070} = \frac{1 - (1.0070)^{-120}}{0.0070} \approx 81.0035 \quad (\text{terlalu besar dibanding } 80.0)$$
- **Tebakan 2:** Karena kita ingin menurunkan nilai dari $81.00$ ke $80.00$, naikkan suku bunganya. Misal $j_2 = 0.73\% = 0.0073$:
  $$a_{\overline{120}\vert 0.0073} = \frac{1 - (1.0073)^{-120}}{0.0073} \approx 79.7601 \quad (\text{terlalu kecil dibanding } 80.0)$$
- **Rumus Interpolasi Linier:**
  $$j \approx j_1 + \frac{PV_{\text{target}} - PV_1}{PV_2 - PV_1} (j_2 - j_1)$$
  $$j \approx 0.0070 + \frac{80.0000 - 81.0035}{79.7601 - 81.0035} (0.0073 - 0.0070)$$
  $$j \approx 0.0070 + \frac{-1.0035}{-1.2434} (0.0003) = 0.0070 + 0.000242 = \mathbf{0.007242 = 0.7242\%}$$

##### Cara 3: Metode Iterasi Newton-Raphson (Komputer / Python)
Menghasilkan nilai eksak berpresisi tinggi:
$$\mathbf{j = 0.00724102 = 0.724102\% \text{ per bulan}}$$

---

### D. Menghitung Suku Bunga Nominal Tahunan ($i^{(12)}$)

Setelah suku bunga bulanan $j$ diperoleh, suku bunga nominal tahunan yang dikonversi bulanan adalah:
$$i^{(12)} = 12 \times j = 12 \times 0.724102\% = \mathbf{8.6892\% \approx 8.69\%}$$

> *(Catatan: Nilai $8.62\%$ pada teks ringkasan adalah hasil pembulatan interpolasi kasar dua titik; nilai eksak yang tepat adalah $\mathbf{8.69\%}$)*.

```python
# Verifikasi Penentuan Suku Bunga Nominal Latihan Soal 2 dengan Python

def a_angle(n, j):
    return (1 - (1 + j)**(-n)) / j

target_pv = 80.0  # a_120|j = 8000 / 100

# 1. Metode Coba-Coba & Interpolasi Linier
j1, j2 = 0.0070, 0.0073
pv1 = a_angle(120, j1)
pv2 = a_angle(120, j2)

j_interpolasi = j1 + (target_pv - pv1) / (pv2 - pv1) * (j2 - j1)
i12_interpolasi = 12 * j_interpolasi

# 2. Metode Iterasi Newton-Raphson (Eksak)
j_exact = 0.007
for _ in range(20):
    f = a_angle(120, j_exact) - target_pv
    df = (120 * (1 + j_exact)**(-121) * j_exact - (1 - (1 + j_exact)**(-120))) / (j_exact**2)
    j_exact -= f / df

i12_exact = 12 * j_exact

print("=== HASIL PERHITUNGAN LATIHAN SOAL 2 ===")
print(f"Target Nilai a_120|j              : {target_pv:.2f}")
print(f"Tebakan 1 (j = 0.70%)            : a_120|0.0070 = {pv1:.4f}")
print(f"Tebakan 2 (j = 0.73%)            : a_120|0.0073 = {pv2:.4f}")
print(f"Hasil Interpolasi Linier (j)     : {j_interpolasi:.6%} per bulan")
print(f"Suku Bunga Nominal i^(12) Interp : {i12_interpolasi:.4%}")
print("-" * 55)
print(f"Hasil Eksak Newton-Raphson (j)   : {j_exact:.6%} per bulan")
print(f"Suku Bunga Nominal i^(12) Eksak  : {i12_exact:.4%}")
```
