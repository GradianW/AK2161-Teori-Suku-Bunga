## 1. Pertanyaan 1: Apakah Soal Slide Halaman 14 Berarti Jangka Waktu 4 Tahun?

### Bunyi Soal (Slide Halaman 14):
> *"Latihan: Hutang $5000 dibayarkan per tahun sebanyak 4 kali dengan sukubunga 8%. Tuliskan jadwal amortisasinya."*

### Pertanyaan Anda:
> *"Pada soal slide halaman 14 ini, apakah berarti 4 tahun?"*

---

### Jawaban & Klarifikasi Mendalam:

**YA, BETUL SEKALI! Maksud soal tersebut adalah pinjaman dilunasi dalam jangka waktu 4 tahun, dengan frekuensi 1 kali pembayaran di setiap akhir tahun ($n = 4$ periode tahunan).**

Berikut adalah 3 alasan dan bukti konklusifnya:

#### Bukti 1: Pola Pasangan Soal dengan Contoh Resmi Dosen di Slide Halaman 12
Soal pada Slide 14 ini adalah **latihan mandiri (*practice exercise*)** yang berpasangan langsung dengan **Contoh di Slide Halaman 12**:
- **Teks Soal di Slide 12:**
  > *"Contoh: Hutang $1000 dibayarkan per tahun sebanyak 4 kali dengan sukubunga 8%."*
- **Penyelesaian Resmi Dosen di Slide 12:**
  $$1000 = R \cdot a_{\overline{n}|i} = R \left( \frac{1 - v^n}{i} \right)$$
  $$R = \frac{1000 \times 0{,}08}{1 - (1{,}08)^{-4}} = 301{,}92$$
  Perhatikan bahwa dosen menggunakan **pangkat $-4$** pada $(1{,}08)^{-4}$ dan suku bunga $0{,}08$ per tahun. Ini membuktikan bahwa:
  - Jangka waktu pinjaman adalah **4 tahun**.
  - Jumlah pembayaran $n = 4$.
  - Tabel jadwal amortisasi yang dibuat dosen di Slide 12 terdiri dari periode $0, 1, 2, 3, 4$ (tahunan).

Karena redaksi kalimat pada Slide 14 identik kata-per-kata (*"Hutang $5000 dibayarkan per tahun sebanyak 4 kali dengan sukubunga 8%"*), hanya pokok hutangnya yang diganti dari $1000 menjadi $5000, maka strukturnya **100% sama**: jangka waktu **4 tahun**, dengan $4$ kali cicilan tahunan.

#### Bukti 2: Makna Kebahasaan (*Language Translation*)
- Frase *"dibayarkan per tahun sebanyak 4 kali"* adalah terjemahan harfiah dosen dari istilah standar aktuaria berbahasa Inggris:
  > *"repaid over 4 years by annual payments"* atau *"paid annually in 4 installments"*.
- Frase *"per tahun"* mengindikasikan frekuensi pembayaran (tahunan), sedangkan *"sebanyak 4 kali"* menyatakan total banyaknya cicilan hingga lunas ($n = 4$).

#### Bukti 3: Konvensi Penulisan Suku Bunga
- Jika dosen bermaksud pembayaran kuartalan (4 kali dalam 1 tahun), dosen **selalu** menuliskan frase nominal: *"convertible quarterly"* (seperti pada Slide 15: *"sukubunga 8% convertible quarterly"* dan Slide 25: *"sukubunga 12% convertible quarterly"*).
- Karena pada Slide 14 hanya tertulis *"sukubunga 8%"*, dalam matematika keuangan hal ini secara baku berarti **suku bunga efektif tahunan** ($i = 8\%$), sehingga periode pemajemukannya adalah tahunan.

```python
# =====================================================================
# SIMULASI & VERIFIKASI SOAL SLIDE 14 (n = 4 TAHUN, i = 8% PER TAHUN)
# =====================================================================
import pandas as pd

L = 5000.0   # Pokok pinjaman
i = 0.08     # Suku bunga tahunan
n = 4        # Durasi: 4 tahun

# 1. Hitung Angsuran Tahunan R
# Formula: R = L / a_angle_4|8%
v_4 = (1 + i) ** (-n)
a_4 = (1 - v_4) / i
R = L / a_4

print(f"Faktor Anuitas a_4|8% = {a_4:.8f}")
print(f"Besar Angsuran R     = ${R:.2f} (Nilai eksak: {R:.6f})")

# 2. Rekonstruksi Tabel Amortisasi Lengkap
schedule = []
B = L
schedule.append({"Periode (t)": 0, "Angsuran (R)": "---", "Bunga (I_t)": "---", "Pokok (P_t)": "---", "Saldo (B_t)": f"${B:,.2f}"})

total_I = 0.0
total_P = 0.0
total_R = 0.0
R_round = round(R, 2)  # $1509.60

for t in range(1, n + 1):
    I_t = round(B * i, 2)
    if t < n:
        P_t = round(R_round - I_t, 2)
        B_new = round(B - P_t, 2)
    else:
        P_t = B
        B_new = 0.0
    
    total_I += I_t
    total_P += P_t
    total_R += R_round
    B = B_new
    
    schedule.append({
        "Periode (t)": t,
        "Angsuran (R)": f"${R_round:,.2f}",
        "Bunga (I_t)": f"${I_t:,.2f}",
        "Pokok (P_t)": f"${P_t:,.2f}",
        "Saldo (B_t)": f"${B_new:,.2f}"
    })

df_schedule = pd.DataFrame(schedule)
print("\n=== TABEL JADWAL AMORTISASI LENGKAP (SLIDE 14) ===")
print(df_schedule.to_string(index=False))
print(f"\nTotal Pembayaran Angsuran = ${total_R:,.2f}")
print(f"Total Bunga yang Dibayar   = ${total_I:,.2f}")
print(f"Total Pokok yang Dilunasi  = ${total_P:,.2f} (Tepat melunasi L = $5000)")
```

## 2. Bedah Soal Slide Halaman 25: Restrukturisasi Suku Bunga Pinjaman Bulanan

### Bunyi Soal:
> *"A meminjam uang sebesar $10,000 ke bank dan akan dikembalikan setiap bulan selama 2 tahun. Sukubunga semula adalah 12% convertible quarterly. Namun terjadi perubahan sukubunga pada awal tahun kedua menjadi 12.4% convertible quarterly. Besar besar angsuran tiap bulan selama 1 tahun adalah 500, berapa besar angsuran untuk 1 tahun kemudian?"*

### Inti Konseptual Pengerjaan:
1. **Tahun 1 (Bulan 1 s.d. 12):**
   - Suku bunga kuartalan awal: $i_q = \frac{12\%}{4} = 3\% = 0{,}03$.
   - Suku bunga efektif bulanan $i_m$ memenuhi $(1 + i_m)^3 = 1{,}03 \implies i_m = (1{,}03)^{1/3} - 1 \approx 0{,}009901634$.
   - Karena cicilan selama 12 bulan pertama ditetapkan $500/bulan, kita gunakan **Metode Retrospektif** untuk mencari saldo sisa pada akhir tahun ke-1 ($B_{12}$):
     $$B_{12} = 10.000(1 + i_m)^{12} - 500 \cdot s_{\overline{12}|i_m}$$
     $$(1 + i_m)^{12} = \left[(1 + i_m)^3\right]^4 = (1{,}03)^4 = 1{,}12550881$$
     $$B_{12} = 11.255{,}09 - 6.337{,}78 = \mathbf{\$4917{,}31}$$

2. **Tahun 2 (Bulan 13 s.d. 24):**
   - Suku bunga kuartalan baru: $j_q = \frac{12{,}4\%}{4} = 3{,}1\% = 0{,}031$.
   - Suku bunga bulanan baru $j_m$: $(1 + j_m)^3 = 1{,}031 \implies j_m = (1{,}031)^{1/3} - 1 \approx 0{,}01022836$.
   - Sisa saldo $B_{12}$ harus lunas dalam 12 bulan sisa dengan angsuran baru $X$:
     $$B_{12} = X \cdot a_{\overline{12}|j_m} \implies X = \frac{4917{,}31}{11{,}237218} \approx \mathbf{\$437{,}59}$$
     *(Secara eksak tanpa pembulatan intermediet: $X = \mathbf{\$437{,}53}$)*.

```python
# =====================================================================
# SIMULASI & VERIFIKASI SOAL SLIDE 25
# =====================================================================
L = 10000.0
R1 = 500.0

# 1. Suku Bunga Tahun 1
i_q = 0.12 / 4        # 3% per kuartal
i_m = (1 + i_q) ** (1/3) - 1  # Bunga efektif bulanan

# Saldo Retrospektif Akhir Tahun 1 (t = 12 bulan)
acc_L = L * ((1 + i_q) ** 4)
s_12 = (((1 + i_m) ** 12) - 1) / i_m
acc_R1 = R1 * s_12
B12 = acc_L - acc_R1

# 2. Suku Bunga Tahun 2
j_q = 0.124 / 4       # 3.1% per kuartal
j_m = (1 + j_q) ** (1/3) - 1  # Bunga efektif bulanan baru

# 3. Angsuran Baru X untuk 12 Bulan Sisa
v_j_12 = (1 + j_q) ** (-4)
a_12_j = (1 - v_j_12) / j_m
X_exact = B12 / a_12_j

print("=== HASIL PERHITUNGAN SOAL SLIDE 25 ===")
print(f"Bunga bulanan tahun 1 (i_m)  = {i_m*100:.5f}%")
print(f"Saldo akhir tahun 1 (B_12)   = ${B12:,.2f}")
print(f"Bunga bulanan tahun 2 (j_m)  = {j_m*100:.5f}%")
print(f"Faktor Anuitas a_12|j_m      = {a_12_j:.6f}")
print(f"Besar Angsuran Baru Tahun 2  = ${X_exact:.2f}")
```

## 3. Pertanyaan 2: Mengapa Saat Mencari Suku Bunga Efektif Dipangkatkan 3? ($(1+i_m)^3 = 1{,}03$)

### Pertanyaan Anda:
> *"saya tidak paham kenapa saat mencari suku bunga efektif dipangkatkan 3"*

---

### Penjelasan Konsep & Logika Finansial:

Angka **3** muncul karena **hubungan waktu antara 1 kuartal (triwulan) dan 1 bulan**:
$$\mathbf{1\text{ Kuartal} = 3\text{ Bulan}}$$

Mari kita bedah langkah demi langkah:

#### 1. Apa yang Diketahui dari Soal?
Suku bunga yang diberikan bank adalah nominal **$12\%$ convertible quarterly** ($i^{(4)} = 12\%$).
- Artinya, bunga dimajemukkan setiap 1 kuartal (setiap 3 bulan).
- Suku bunga untuk 1 kuartal ($i_q$) adalah:
  $$i_q = \frac{12\%}{4} = 3\% = 0{,}03\text{ per kuartal}$$
- Jadi, jika kita menyimpan uang sebesar $1 di awal kuartal, maka setelah **1 kuartal (yaitu 3 bulan)**, uang kita tumbuh menjadi:
  $$\$1 \times (1 + i_q) = 1 + 0{,}03 = \mathbf{1{,}03}$$

#### 2. Apa yang Dibutuhkan oleh Soal?
Karena debitur membayar angsuran **setiap bulan**, kita wajib mengetahui **suku bunga efektif bulanan ($i_m$)** agar rumus anuitas bulanan dapat digunakan.

#### 3. Bagaimana Uang Tumbuh Secara Bulanan?
Misalkan suku bunga per bulan adalah $i_m$. Jika uang $1 dibungakan secara bulanan:
- Setelah **1 bulan**: menjadi $(1 + i_m)$
- Setelah **2 bulan**: menjadi $(1 + i_m) \times (1 + i_m) = (1 + i_m)^2$
- Setelah **3 bulan**: menjadi $(1 + i_m)^2 \times (1 + i_m) = \mathbf{(1 + i_m)^3}$

#### 4. Prinsip Ekuivalensi Finansial (Titik Temunya):
Karena **3 bulan itu sama persis dengan 1 kuartal**, maka uang $1 yang dibungakan secara bulanan selama 3 bulan **harus bernilai sama persis** dengan uang $1 yang dibungakan secara kuartalan selama 1 kuartal:
$$\underbrace{(1 + i_m)^3}_{\text{Pertumbuhan selama 3 bulan}} = \underbrace{1 + i_q}_{\text{Pertumbuhan selama 1 kuartal}} = \mathbf{1{,}03}$$

#### 5. Cara Mendapatkan $i_m$ (Menarik Akar Pangkat 3):
Untuk mencari $i_m$, kita pangkatkan kedua ruas dengan $\frac{1}{3}$ (atau tarik akar pangkat 3):
$$\left[ (1 + i_m)^3 \right]^{1/3} = (1{,}03)^{1/3}$$
$$1 + i_m = (1{,}03)^{1/3}$$
$$i_m = (1{,}03)^{1/3} - 1 \approx 1{,}009901634 - 1 = \mathbf{0{,}009901634} \quad (\approx 0{,}99016\%\text{ per bulan})$$

---

#### Sudut Pandang Alternatif: Menyamakan Horison Waktu 1 Tahun Penuh
Dalam 1 tahun:
- Terdapat **12 periode bulanan** $\implies$ akumulasi tahunan $= (1 + i_m)^{12}$
- Terdapat **4 periode kuartalan** $\implies$ akumulasi tahunan $= (1 + i_q)^4 = (1{,}03)^4$

Samakan akumulasi 1 tahun keduanya:
$$(1 + i_m)^{12} = (1{,}03)^4$$
Pangkatkan kedua ruas dengan $\frac{1}{4}$:
$$\left[ (1 + i_m)^{12} \right]^{1/4} = \left[ (1{,}03)^4 \right]^{1/4} \implies \mathbf{(1 + i_m)^3 = 1{,}03}$$
Hasilnya identik! Angka **3** berasal dari rasio $\frac{12\text{ bulan}}{4\text{ kuartal}} = 3\text{ bulan/kuartal}$.

```python
# Verifikasi Ekuivalensi Pertumbuhan Nilai Uang (Bulanan vs Kuartalan)
i_q = 0.03
i_m = (1 + i_q)**(1/3) - 1

print(f"Bunga kuartalan i_q   = {i_q*100:.2f}%")
print(f"Bunga bulanan i_m     = {i_m*100:.5f}%")
print(f"Akumulasi 3 bulan     = (1 + i_m)^3 = {(1 + i_m)**3:.4f}")
print(f"Akumulasi 1 kuartal   = 1 + i_q     = {1 + i_q:.4f}")
print(f"Apakah keduanya sama persis? {abs((1 + i_m)**3 - (1 + i_q)) < 1e-12}")
```

## 4. Pertanyaan 3: Kenapa Bukan $1+i = \left(1 + \frac{i^{(4)}}{4}\right)^4$?

### Pertanyaan Anda:
> *"kenapa bukan $1+i = (1+i^{(4)}/4)^4$?"*

---

### Jawaban & Klarifikasi Sangat Penting:

**Rumus Anda $1 + i = \left( 1 + \frac{i^{(4)}}{4} \right)^4$ itu 100% BENAR!**

Tidak ada yang salah dengan rumus tersebut. Hanya saja, mari kita perhatikan **satuan waktu** dari simbol $i$ pada rumus Anda:
- Simbol $i$ pada rumus tersebut adalah **suku bunga efektif TAHUNAN** (*annual effective rate*).
- $1 + i = (1 + 0{,}03)^4 = (1{,}03)^4 \approx 1{,}12550881 \implies i \approx 12{,}55088\%\text{ per tahun}$.

#### Masalahnya: Apakah Cicilan Dibayar 1 Tahun Sekali?
- Di soal ini, debitur membayar cicilan **setiap BULAN** ($R_1 = \$500$ per bulan selama 12 bulan).
- Jika kita ingin menghitung nilai akumulasi cicilan bulanan $s_{\overline{12}|}$, kita **tidak boleh** menggunakan suku bunga tahunan $i$, melainkan harus menggunakan **suku bunga efektif BULANAN ($i_m$)**.

#### Menghubungkan Rumus Anda dengan Suku Bunga Bulanan:
Karena dalam 1 tahun ada 12 bulan, pertumbuhan uang selama 1 tahun dapat ditulis dengan suku bunga bulanan $i_m$ sebagai:
$$1 + i = (1 + i_m)^{12}$$

Sekarang, gabungkan rumus Anda dengan rumus bulanan ini:
$$\underbrace{(1 + i_m)^{12}}_{\text{1 tahun via bulanan}} = \underbrace{1 + i}_{\text{1 tahun tahunan}} = \underbrace{\left(1 + \frac{i^{(4)}}{4}\right)^4}_{\text{1 tahun via kuartalan}}$$

Perhatikan hubungan ujung kiri dan ujung kanan:
$$(1 + i_m)^{12} = (1{,}03)^4$$

Jika Anda ingin mencari $i_m$, tarik akar pangkat 12 (pangkatkan kedua ruas dengan $\frac{1}{12}$):
$$1 + i_m = \left[ (1{,}03)^4 \right]^{\frac{1}{12}} = (1{,}03)^{\frac{4}{12}} = \mathbf{(1{,}03)^{\frac{1}{3}}}$$

Lihat! Dari rumus Anda pun, pangkat $\frac{4}{12}$ disederhanakan menjadi **$\frac{1}{3}$**!

#### Kesimpulan:
1. Rumus Anda $1 + i = (1 + i^{(4)}/4)^4$ mengonversi dari kuartalan ke **tahunan**.
2. Rumus $(1 + i_m)^3 = 1{,}03$ mengonversi dari kuartalan langsung ke **bulanan**.
3. Keduanya identik dan menghasilkan nilai suku bunga yang persis sama!

```python
# Pembuktian: Jalur Rumus Anda (Lewat Tahunan) vs Jalur Langsung (Bulanan)
i_q = 0.03

# Jalur Rumus Anda: Cari i tahunan dulu, lalu cari i_m
one_plus_i_annual = (1 + i_q) ** 4          # 1 + i = (1 + 0.03)^4
i_annual = one_plus_i_annual - 1
i_m_from_annual = (one_plus_i_annual) ** (1/12) - 1

# Jalur Langsung: (1 + i_m)^3 = 1 + i_q
i_m_direct = (1 + i_q) ** (1/3) - 1

print(f"Suku bunga efektif tahunan i = {i_annual*100:.6f}%")
print(f"i_m via rumus tahunan Anda   = {i_m_from_annual*100:.6f}%")
print(f"i_m via rumus langsung       = {i_m_direct*100:.6f}%")
print(f"Selisih keduanya             = {abs(i_m_from_annual - i_m_direct):.2e} (IDENTIK!)")
```
