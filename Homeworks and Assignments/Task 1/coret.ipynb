## 1. Pembahasan Soal 1: Hubungan Nominal Rate $i^{(m)}$ dan $d^{(m)}$

### Bunyi Soal
> Bila diketahui $i^{(m)} = a$ dan $d^{(m)} = b$, tentukan $m$ dalam $a$ dan $b$.

---

### Penurunan Matematis

1. **Definisi Suku Bunga Nominal ($i^{(m)}$) dan Tingkat Diskon Nominal ($d^{(m)}$):**
   $$1 + i = \left( 1 + \frac{i^{(m)}}{m} \right)^m = \left( 1 + \frac{a}{m} \right)^m$$
   $$1 - d = \left( 1 - \frac{d^{(m)}}{m} \right)^m = \left( 1 - \frac{b}{m} \right)^m$$

2. **Relasi Fundamental Teori Suku Bunga:**
   $$(1+i)(1-d) = 1 \iff (1+i) = \frac{1}{1-d}$$

3. **Substitusi dan Eliminasi Pangkat $m$:**
   $$\left( 1 + \frac{a}{m} \right)^m \left( 1 - \frac{b}{m} \right)^m = 1$$
   Tarik akar ke-$m$ di kedua ruas (karena $1 + a/m > 0$ dan $1 - b/m > 0$):
   $$\left( 1 + \frac{a}{m} \right) \left( 1 - \frac{b}{m} \right) = 1$$

4. **Penjabaran Aljabar:**
   $$1 - \frac{b}{m} + \frac{a}{m} - \frac{ab}{m^2} = 1$$
   $$\frac{a - b}{m} - \frac{ab}{m^2} = 0$$

5. **Selesaikan untuk $m$ ($m \neq 0$):**
   Kalikan kedua ruas dengan $m^2$:
   $$(a - b)m - ab = 0 \implies (a - b)m = ab \implies \mathbf{m = \frac{ab}{a - b}}$$

> **Cara Cepat Alternatif (Identitas Resiprokal):**
> $$\frac{1}{d^{(m)}} - \frac{1}{i^{(m)}} = \frac{1}{m} \implies \frac{1}{b} - \frac{1}{a} = \frac{1}{m} \implies \frac{a - b}{ab} = \frac{1}{m} \implies \mathbf{m = \frac{ab}{a - b}}$$

```python
# Verifikasi Hubungan m = ab / (a - b) dengan Python

i_eff = 0.10  # Misal suku bunga tahunan 10%
m_actual = 12  # Misal frekuensi konversi bulanan

# Hitung nominal rate a = i^(12) dan b = d^(12)
a = m_actual * ((1 + i_eff)**(1/m_actual) - 1)
b = m_actual * (1 - (1 + i_eff)**(-1/m_actual))

# Hitung kembali m menggunakan formula
m_calculated = (a * b) / (a - b)

print(f"Nilai a = i^(12) : {a:.6f}")
print(f"Nilai b = d^(12) : {b:.6f}")
print(f"m Aktual       : {m_actual}")
print(f"m Terhitung    : {m_calculated:.4f}")
```

## 2. Pembahasan Soal 3: Evaluasi Investasi Projek A vs Projek B

### Bunyi Soal
> Irfan memiliki modal sebesar 1 juta Rupiah. Ia memiliki 2 opsi untuk melakukan investasi yaitu ikut dalam Projek A atau Projek B. Kedua projek tersebut memerlukan modal yang sama yaitu 1 juta Rupiah.
> - Jika berinvestasi pada **Projek A**, Irfan akan mendapatkan **1.5 juta Rupiah di akhir tahun pertama**.
> - Sedangkan **Projek B** akan memberikan cashflow sebesar **750 ribu Rupiah di akhir tahun pertama** dan **1 juta Rupiah di akhir tahun kedua**.
> Tentukan IRR Projek A dan Projek B, kemudian beri keputusan sebaiknya Irfan melakukan investasi pada Projek A atau B!

---

### Penurunan Matematis IRR (*Internal Rate of Return*)

#### A. Projek A
- Modal awal: $C_0 = -1\text{ juta}$ pada $t = 0$.
- Penerimaan: $C_1 = +1.5\text{ juta}$ pada $t = 1$.
- Persamaan Nilai Bersih Sekarang ($NPV = 0$):
  $$NPV_A(r_A) = -1 + \frac{1.5}{1 + r_A} = 0$$
  $$1 + r_A = 1.5 \implies \mathbf{r_A = 0.50 = 50\%}$$

---

#### B. Projek B
- Modal awal: $C_0 = -1\text{ juta}$ pada $t = 0$.
- Penerimaan 1: $C_1 = +0.75\text{ juta}$ pada $t = 1$.
- Penerimaan 2: $C_2 = +1.0\text{ juta}$ pada $t = 2$.
- Persamaan Nilai Bersih Sekarang ($NPV = 0$):
  $$NPV_B(r_B) = -1 + \frac{0.75}{1 + r_B} + \frac{1}{(1 + r_B)^2} = 0$$

Misalkan faktor diskonto $v = \frac{1}{1 + r_B}$:
$$-1 + 0.75v + v^2 = 0 \iff v^2 + \frac{3}{4}v - 1 = 0 \iff 4v^2 + 3v - 4 = 0$$

Gunakan rumus kuadrat abc:
$$v_{1,2} = \frac{-3 \pm \sqrt{3^2 - 4(4)(-4)}}{2(4)} = \frac{-3 \pm \sqrt{73}}{8}$$

Terdapat **dua kemungkinan nilai $v$**:
1. **Akar Pertama ($v_1$, Positif / Valid Secara Finansial):**
   $$v_1 = \frac{-3 + \sqrt{73}}{8} \approx \frac{-3 + 8.5440037}{8} = \frac{5.5440037}{8} \approx \mathbf{0.6930005}$$
   Maka suku bunga $r_B$ adalah:
   $$1 + r_B = \frac{1}{v_1} = \frac{8}{\sqrt{73} - 3} = \frac{\sqrt{73} + 3}{8} \approx 1.4430005 \implies \mathbf{r_B = 44.30\%}$$

2. **Akar Kedua ($v_2$, Negatif / Ditolak Secara Finansial):**
   $$v_2 = \frac{-3 - \sqrt{73}}{8} \approx \frac{-3 - 8.5440037}{8} = \frac{-11.5440037}{8} \approx \mathbf{-1.4430005}$$
   - Jika dihitung nilai $r$-nya: $1 + r = \frac{1}{v_2} = -0.6930005 \implies r = -169.30\%$.
   - **Ditolak (*Extraneous Root*)** karena faktor diskonto tidak boleh bernilai negatif ($v = (1+r)^{-1} > 0$) dan tingkat bunga $r < -100\%$ tidak memiliki arti riil dalam studi kelayakan proyek.

---

#### C. Keputusan Investasi
$$\mathbf{IRR_A \, (50\%) > IRR_B \, (44.30\%)}$$

Karena kedua projek memerlukan modal awal yang sama besar ($\$1\text{ juta}$), maka berdasarkan kriteria IRR, Irfan sebaiknya memilih **Projek A**.

```python
# Verifikasi IRR Projek A dan Projek B dengan Python
import numpy as np
import numpy_financial as npf

# Cashflows
cf_A = [-1.0, 1.5]
cf_B = [-1.0, 0.75, 1.0]

irr_A = npf.irr(cf_A)
irr_B = npf.irr(cf_B)

print("=== HASIL PERHITUNGAN IRR ===")
print(f"IRR Projek A : {irr_A:.4%}")
print(f"IRR Projek B : {irr_B:.4%}")
print(f"Rekomendasi  : {'Pilih Projek A' if irr_A > irr_B else 'Pilih Projek B'}")
```

## 3. Klarifikasi Konsep: Metode Apa yang Digunakan pada Soal 3? (*Dollar-Weighted* vs *Time-Weighted*)

### Pertanyaan
> *"Metode apa yang kamu pakai untuk menjawab soal 3.? Apakah dollar weighted atau time weighted??"*

---

### A. Jawaban Langsung

Metode yang digunakan untuk menjawab Soal 3 adalah **DOLLAR-WEIGHTED RATE OF RETURN (DWRR)**, atau yang secara universal di dunia keuangan dikenal sebagai **IRR (*Internal Rate of Return*) / *Yield Rate***.

Metode **Time-Weighted Rate of Return (TWRR) TIDAK BISA DIGUNAKAN** pada soal ini.

---

### B. Mengapa IRR = Dollar-Weighted Rate of Return (DWRR)?

Dalam literatur resmi aktuaria (*Theory of Interest* - Kellison Chapter 5 / Vaaler & Daniel Chapter 4):
1. **Definisi *Dollar-Weighted Rate of Return* (DWRR):**
   DWRR adalah tingkat suku bunga majemuk $r$ yang menyamakan nilai sekarang seluruh kas masuk (*cash inflows*) dengan nilai sekarang seluruh kas keluar (*cash outflows*):
   $$\sum_{t} C_t (1 + r)^{-t} = 0$$
   Persamaan ini tidak lain adalah **Persamaan Nilai Waktu dari Uang (*Equation of Value*) / IRR**.
2. *Catatan Tambahan:* Terkadang di materi dasar, DWRR diajarkan dalam bentuk aproksimasi bunga sederhana (*simple interest approximation*), yaitu $i \approx \frac{I}{A + \sum C_t (1 - t)}$ untuk rentang waktu $1$ tahun. Namun untuk rentang multi-tahun atau proyek investasi, bentuk eksaknya adalah **IRR**.

---

### C. Mengapa *Time-Weighted Rate of Return* (TWRR) Tidak Bisa Dipakai pada Soal 3?

1. **Kebutuhan Data TWRR:**
   Untuk menghitung TWRR, kita **WAJIB mengetahui nilai pasar / saldo portofolio investasi ($B_t$) tepat sebelum arus kas masuk/keluar terjadi**.
   Rumus TWRR adalah:
   $$1 + r_{\text{TWRR}} = \left( \frac{B_1}{B_0} \right) \times \left( \frac{B_2}{B_1 + C_1} \right) \times \dots$$

2. **Kondisi Soal 3:**
   Pada Soal 3, kita **hanya diberikan arus kas proyek** (modal $\$1\text{ juta}$ di awal, hasil $\$750\text{ ribu}$ di tahun 1, dan $\$1\text{ juta}$ di tahun 2). Kita **TIDAK PERNAH DIBERI TAHU** berapa saldo nilai pasar aset (*market value*) proyek B pada saat $t = 1$ sebelum uang $\$750\text{ ribu}$ diambil.

3. **Tujuan Finansial:**
   - **TWRR** dirancang untuk mengevaluasi **kinerja manajer investasi (reksa dana/portofolio saham)** agar tidak bias oleh tindakan nasabah yang menyetor/menarik uang.
   - **DWRR / IRR** dirancang untuk mengevaluasi **keputusan proyek penganggaran modal (*capital budgeting*)**, yaitu apakah suatu proyek bisnis menghasilkan tingkat pengembalian uang yang memadai.

---

### D. Tabel Perbandingan Kontras: DWRR (IRR) vs TWRR

| Karakteristik | *Dollar-Weighted Rate of Return* (DWRR / IRR) | *Time-Weighted Rate of Return* (TWRR) |
| :--- | :--- | :--- |
| **Konsep Dasar** | Menghitung tingkat bunga efektif yang membuat $NPV = 0$. | Menghitung rata-rata geometrik dari tingkat imbal hasil setiap sub-periode. |
| **Data yang Dibutuhkan** | Hanya data **arus kas ($C_t$)** dan **titik waktu ($t$)**. | Data **arus kas ($C_t$)** DAN **saldo nilai pasar portofolio ($B_t$) sebelum arus kas**. |
| **Pengaruh Waktu Arus Kas** | **Sangat dipengaruhi** oleh besar dan waktu penarikan uang. | **Kebal / Tidak terpengaruh** oleh waktu dan besar arus kas masuk/keluar nasabah. |
| **Penggunaan Utama** | Studi kelayakan proyek modal, investasi bisnis, asuransi, pinjaman. | Evaluasi kinerja manajer dana investasi (*fund manager* / reksa dana). |
| **Penerapan pada Soal 3** | ✅ **Sangat tepat dan merupakan satu-satunya metode yang valid**. | ❌ **Tidak bisa diterapkan (karena data nilai pasar $B_1$ tidak ada)**. |

## 4. Pembahasan Soal 5: Present Value dari 10-Year Annuity Due dengan Bunga Berubah

### Bunyi Soal
> Tentukan present value dari **10-year annuity due** dari pembayaran **$\$1$ per tahun**, jika suku bunga efektif **$4\%$ untuk 5 tahun pertama** dan **$5\%$ untuk 5 tahun berikutnya**!

---

### Garis Waktu Arus Kas (*Timeline*)

Karena transaksi adalah **Annuity-Due** (pembayaran di awal tahun), pembayaran $\$1$ terjadi pada titik waktu $t = 0, 1, 2, \dots, 9$:

| Rentang Waktu | 5 Tahun Pertama (Tahun 1 s.d. 5) | 5 Tahun Kedua (Tahun 6 s.d. 10) |
| :--- | :---: | :---: |
| **Titik Waktu Pembayaran ($t$)** | $t = 0, 1, 2, 3, 4$ | $t = 5, 6, 7, 8, 9$ |
| **Besar Pembayaran** | $\$1$ per awal tahun | $\$1$ per awal tahun |
| **Suku Bunga Efektif** | $i_1 = 4\% = 0.04$ | $i_2 = 5\% = 0.05$ |
| **Faktor Diskonto** | $v_1 = (1.04)^{-1}$ | $v_2 = (1.05)^{-1}$ |

---

### Penurunan Matematis Nilai Sekarang ($PV$ di $t = 0$)

Nilai sekarang adalah jumlahan diskonto dari kedua blok pembayaran:
$$PV = \underbrace{\sum_{t=0}^4 v_1^t}_{\text{Blok 1 (5 bayar pertama di } t=0,1,2,3,4)} + \underbrace{v_1^5 \sum_{k=0}^4 v_2^k}_{\text{Blok 2 (5 bayar kedua di } t=5,6,7,8,9)}$$

Dalam notasi aktuarial:
$$PV = \ddot{a}_{\overline{5}\vert 0.04} + v_1^5 \cdot \ddot{a}_{\overline{5}\vert 0.05}$$

#### Perhitungan Komponen:
1. **Blok 1 (5 Pembayaran Pertama):**
   $$\ddot{a}_{\overline{5}\vert 0.04} = (1.04) \cdot \left[ \frac{1 - (1.04)^{-5}}{0.04} \right] = (1.04) \times 4.451822 = \mathbf{4.629895}$$

2. **Blok 2 (5 Pembayaran Kedua di Titik $t = 5$):**
   $$\ddot{a}_{\overline{5}\vert 0.05} = (1.05) \cdot \left[ \frac{1 - (1.05)^{-5}}{0.05} \right] = (1.05) \times 4.329477 = \mathbf{4.545951}$$

3. **Diskontokan Blok 2 dari $t = 5$ ke $t = 0$ (menggunakan bunga 5 tahun pertama):**
   $$v_1^5 \cdot \ddot{a}_{\overline{5}\vert 0.05} = (1.04)^{-5} \times 4.545951 = 0.8219271 \times 4.545951 = \mathbf{3.736440}$$

4. **Total Nilai Sekarang ($PV$):**
   $$PV = 4.629895 + 3.736440 = \mathbf{8.366335 \approx \$8.37}$$

```python
# Verifikasi PV Soal 5 dengan Python

i1, i2 = 0.04, 0.05
v1 = 1 / (1 + i1)
v2 = 1 / (1 + i2)

# 1. Perhitungan langsung suku per suku
pv_suku_per_suku = (
    1.0
    + v1**1
    + v1**2
    + v1**3
    + v1**4
    + (v1**5) * 1.0
    + (v1**5) * (v2**1)
    + (v1**5) * (v2**2)
    + (v1**5) * (v2**3)
    + (v1**5) * (v2**4)
)

# 2. Perhitungan menggunakan formula anuitas
a_due_1 = (1 + i1) * ((1 - v1**5) / i1)
a_due_2 = (1 + i2) * ((1 - v2**5) / i2)
pv_formula = a_due_1 + (v1**5) * a_due_2

print("=== HASIL VALUASI NILAI SEKARANG (PV) SOAL 5 ===")
print(f"a_due 5 (i=4%)           : {a_due_1:.6f}")
print(f"a_due 5 (i=5%)           : {a_due_2:.6f}")
print(f"Diskonto Blok 2 ke t=0   : {(v1**5) * a_due_2:.6f}")
print("-" * 50)
print(f"PV (Suku per Suku)       : {pv_suku_per_suku:.6f}")
print(f"PV (Formula Anuitas)     : {pv_formula:.6f}")
```
