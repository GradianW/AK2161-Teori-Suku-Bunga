```python
# Definisi Fungsi Polinomial P(v) dan f(i)
import numpy as np
import pandas as pd

def P(v):
    return 12*v**10 + 10*v**9 + 9*v**8 + 8*v**7 + 7*v**6 - v**5 - v**4 - v**3 - v**2 - 5*v - 10

def P_prime(v):
    return 120*v**9 + 90*v**8 + 72*v**7 + 56*v**6 + 42*v**5 - 5*v**4 - 4*v**3 - 3*v**2 - 2*v - 5

def f(i):
    return P(1.0 / (1.0 + i))

# 1. Newton-Raphson (13 Iterasi)
hist_nr = []
v = 0.90
for k in range(1, 14):
    pv = P(v)
    ppv = P_prime(v)
    v_next = v - pv/ppv
    i_k = 1.0/v_next - 1.0
    hist_nr.append({'Iterasi (k)': k, 'v_k': f'{v:.15f}', 'P(v_k)': f'{pv:+.15f}', "P'(v_k)": f'{ppv:.15f}', 'Estimasi i_k': f'{i_k*100:.15f}%'})
    v = v_next
df_nr = pd.DataFrame(hist_nr)

# 2. Secant Method (13 Iterasi)
hist_sec = []
x0, x1 = 0.10, 0.15
for k in range(1, 14):
    fx0, fx1 = f(x0), f(x1)
    x2 = x1 - fx1 * (x1 - x0) / (fx1 - fx0) if abs(fx1 - fx0) > 1e-15 else x1
    hist_sec.append({'Iterasi (k)': k, 'i_{k-1}': f'{x0:.15f}', 'i_k': f'{x1:.15f}', 'f(i_k)': f'{fx1:+.15f}', 'Estimasi i_{k+1}': f'{x2*100:.15f}%'})
    x0, x1 = x1, x2
df_sec = pd.DataFrame(hist_sec)

# 3. Regula Falsi (13 Iterasi)
hist_rf = []
a, b = 0.10, 0.15
for k in range(1, 14):
    fa, fb = f(a), f(b)
    c = (a*fb - b*fa) / (fb - fa)
    fc = f(c)
    hist_rf.append({'Iterasi (k)': k, 'Batas Kiri a_k': f'{a:.15f}', 'Batas Kanan b_k': f'{b:.15f}', 'Estimasi c_k': f'{c*100:.15f}%', 'f(c_k)': f'{fc:+.15f}'})
    if fa * fc < 0:
        b = c
    else:
        a = c
df_rf = pd.DataFrame(hist_rf)

# 4. Bisection (13 Iterasi)
hist_bis = []
a, b = 0.10, 0.15
for k in range(1, 14):
    c = (a + b) / 2.0
    fc = f(c)
    hist_bis.append({'Iterasi (k)': k, 'Batas Kiri a_k': f'{a:.15f}', 'Batas Kanan b_k': f'{b:.15f}', 'Titik Tengah c_k': f'{c*100:.15f}%', 'f(c_k)': f'{fc:+.15f}'})
    if f(a) * fc < 0:
        b = c
    else:
        a = c
df_bis = pd.DataFrame(hist_bis)

print("=== 1. NEWTON-RAPHSON (13 ITERASI) ===")
print(df_nr.to_string(index=False))

print("\n=== 2. SECANT (13 ITERASI) ===")
print(df_sec.to_string(index=False))

print("\n=== 3. REGULA FALSI (13 ITERASI) ===")
print(df_rf.to_string(index=False))

print("\n=== 4. BISECTION (13 ITERASI) ===")
print(df_bis.to_string(index=False))
```
